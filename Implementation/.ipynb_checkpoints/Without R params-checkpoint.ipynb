```python
"""
Deep Learning Community Detection Testing
For Master's Research Project
Author: MOUDDEN Hamza

Methods: GCN, GAT, GAN, Autoencoder
Metrics: Modularity, NMI, ARI
"""

import os
import time
import numpy as np
import pandas as pd
import networkx as nx
from collections import defaultdict
from typing import Dict
import ast
import warnings
warnings.filterwarnings('ignore')

# TensorFlow/Keras imports
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# ============================================================
# 1. PATH HANDLING AND DATA LOADING
# ============================================================

def get_data_path(filename: str) -> str:
    """Get full path to a dataset in the data folder"""
    possible_paths = [
        os.path.join("data", filename),
        os.path.join("../data", filename),
        os.path.join(os.getcwd(), "data", filename),
        os.path.join(os.path.dirname(os.getcwd()), "data", filename),
        filename
    ]
    
    for path in possible_paths:
        if os.path.exists(path):
            return path
    
    return os.path.join("data", filename)

def load_ground_truth_from_txt(gt_filename: str) -> Dict[int, int]:
    """Load ground truth communities from a text file"""
    ground_truth = {}
    
    if not gt_filename.endswith('.txt'):
        gt_filename = gt_filename + '.txt'
    
    path = get_data_path(gt_filename)
    
    if not os.path.exists(path):
        return ground_truth
    
    try:
        with open(path, 'r') as f:
            content = f.read().strip()
        
        if content.startswith('[') and content.endswith(']'):
            try:
                labels_list = ast.literal_eval(content)
                unique_labels = {}
                next_id = 0
                for node_idx, label in enumerate(labels_list):
                    if label not in unique_labels:
                        unique_labels[label] = next_id
                        next_id += 1
                    ground_truth[node_idx] = unique_labels[label]
            except:
                content = content[1:-1]
                parts = [p.strip().strip('"').strip("'") for p in content.split(',')]
                unique_labels = {}
                next_id = 0
                for node_idx, label in enumerate(parts):
                    if label not in unique_labels:
                        unique_labels[label] = next_id
                        next_id += 1
                    ground_truth[node_idx] = unique_labels[label]
        else:
            lines = content.split('\n')
            for line in lines:
                line = line.strip()
                if line and not line.startswith('#'):
                    parts = line.split()
                    if len(parts) >= 2:
                        try:
                            node = int(parts[0])
                            community = int(parts[1])
                            ground_truth[node] = community
                        except:
                            pass
    
    except Exception as e:
        print(f"  Warning: Could not load ground truth: {e}")
    
    return ground_truth

def load_graph_safe(filepath: str) -> nx.Graph:
    """Safely load a GML file and return an undirected graph"""
    try:
        G = nx.read_gml(filepath)
    except:
        try:
            G = nx.read_gml(filepath, label='id')
        except:
            try:
                G = nx.read_edgelist(filepath)
            except Exception as e:
                raise Exception(f"Cannot load graph: {e}")
    
    if nx.is_directed(G):
        G = G.to_undirected()
    
    G.remove_edges_from(nx.selfloop_edges(G))
    
    node_mapping = {node: i for i, node in enumerate(sorted(G.nodes()))}
    G = nx.relabel_nodes(G, node_mapping)
    
    return G

# ============================================================
# 2. DATASET LOADING FUNCTIONS
# ============================================================

def load_karate_club():
    path = get_data_path("karate.gml")
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("karate_GR")
    if not gt:
        for node in G.nodes():
            gt[node] = 0 if node != 33 else 1
    return G, gt

def load_dolphins():
    path = get_data_path("Dolphins.gml")
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("dolphins_GR")
    return G, gt

def load_football():
    path = get_data_path("Football.gml")
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("football_GR")
    return G, gt

def load_polbooks():
    path = get_data_path("Polbooks.gml")
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("polbooks_GT")
    return G, gt

def load_primary_school():
    path = get_data_path("PS.gml")
    if not os.path.exists(path):
        path = get_data_path("Thiers.gml")
    
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("Thiers_GR")
    if not gt:
        gt = load_ground_truth_from_txt("PSD1_GR")
    if not gt:
        gt = load_ground_truth_from_txt("PSD2_GR")
    
    if not gt:
        for node in G.nodes():
            gt[node] = node % 10
    
    return G, gt

def load_vs13():
    path = get_data_path("VS13.gml")
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("VS13_GR")
    return G, gt

def load_vs15():
    path = get_data_path("VS15.gml")
    G = load_graph_safe(path)
    gt = load_ground_truth_from_txt("VS15_GR")
    return G, gt

# ============================================================
# 3. UTILITY FUNCTIONS
# ============================================================

def get_node_features(G: nx.Graph) -> np.ndarray:
    """Extract node features from graph structure"""
    n = G.number_of_nodes()
    A = nx.to_numpy_array(G)
    
    # Features: normalized adjacency row + degree + clustering coefficient
    degree = np.sum(A, axis=1).reshape(-1, 1)
    degree_norm = degree / (n - 1) if n > 1 else degree
    
    # Clustering coefficient
    clustering = np.array(list(nx.clustering(G).values())).reshape(-1, 1)
    
    # Combine features
    features = np.hstack([A, degree_norm, clustering])
    
    # Reduce dimensionality using PCA-like approach (simple scaling)
    scaler = StandardScaler()
    features = scaler.fit_transform(features)
    
    return features.astype(np.float32)

def calculate_modularity(G: nx.Graph, partition: Dict[int, int]) -> float:
    """Calculate modularity score"""
    m = G.number_of_edges()
    if m == 0:
        return 0.0
    
    communities = defaultdict(set)
    for node, comm in partition.items():
        communities[comm].add(node)
    
    modularity = 0.0
    for comm_nodes in communities.values():
        subgraph = G.subgraph(comm_nodes)
        l_c = subgraph.number_of_edges()
        d_c = sum(G.degree(node) for node in comm_nodes)
        modularity += (l_c / m) - (d_c / (2 * m)) ** 2
    
    return modularity

def calculate_nmi(partition1: Dict[int, int], partition2: Dict[int, int]) -> float:
    """Calculate Normalized Mutual Information"""
    nodes = sorted(set(partition1.keys()) & set(partition2.keys()))
    if not nodes:
        return 0.0
    
    labels1 = [partition1[node] for node in nodes]
    labels2 = [partition2[node] for node in nodes]
    
    from sklearn.metrics import normalized_mutual_info_score
    return normalized_mutual_info_score(labels1, labels2)

def calculate_ari(partition1: Dict[int, int], partition2: Dict[int, int]) -> float:
    """Calculate Adjusted Rand Index"""
    nodes = sorted(set(partition1.keys()) & set(partition2.keys()))
    if not nodes:
        return 0.0
    
    labels1 = [partition1[node] for node in nodes]
    labels2 = [partition2[node] for node in nodes]
    
    from sklearn.metrics import adjusted_rand_score
    return adjusted_rand_score(labels1, labels2)

def embeddings_to_partition(embeddings: np.ndarray, n_communities: int) -> Dict[int, int]:
    """Convert node embeddings to community partition using k-means"""
    if n_communities > len(embeddings):
        n_communities = max(2, len(embeddings) // 5)
    
    kmeans = KMeans(n_clusters=n_communities, random_state=42, n_init=10)
    labels = kmeans.fit_predict(embeddings)
    
    return {i: int(labels[i]) for i in range(len(labels))}

# ============================================================
# 4. GCN MODEL (Simplified)
# ============================================================

def gcn_community_detection(G: nx.Graph, epochs=100, embedding_dim=16) -> Dict[int, int]:
    """Simplified GCN using dense layers with adjacency weighting"""
    print("  Running GCN...")
    
    n = G.number_of_nodes()
    features = get_node_features(G)
    
    # Build model
    inputs = keras.Input(shape=(features.shape[1],))
    x = layers.Dense(64, activation='relu')(inputs)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(32, activation='relu')(x)
    embeddings = layers.Dense(embedding_dim, activation='relu', name='embedding')(x)
    
    # Reconstruction output
    outputs = layers.Dense(features.shape[1], activation='sigmoid')(embeddings)
    
    model = keras.Model(inputs=inputs, outputs=outputs)
    model.compile(optimizer=Adam(learning_rate=0.01), loss='mse')
    
    # Train
    model.fit(features, features, epochs=epochs, batch_size=min(32, n), 
              verbose=0, shuffle=True)
    
    # Get embeddings
    embedding_model = keras.Model(inputs=inputs, outputs=embeddings)
    node_embeddings = embedding_model.predict(features, verbose=0)
    
    n_communities = max(2, int(np.sqrt(n)))
    partition = embeddings_to_partition(node_embeddings, n_communities)
    
    return partition

# ============================================================
# 5. GAT MODEL (Simplified with attention-like mechanism)
# ============================================================

def gat_community_detection(G: nx.Graph, epochs=100, embedding_dim=16) -> Dict[int, int]:
    """Simplified GAT using multi-head attention style"""
    print("  Running GAT...")
    
    n = G.number_of_nodes()
    features = get_node_features(G)
    A = nx.to_numpy_array(G)
    
    # Build model with attention-like mechanism
    inputs = keras.Input(shape=(features.shape[1],))
    
    # First layer
    x = layers.Dense(64, activation='relu')(inputs)
    x = layers.Dropout(0.3)(x)
    
    # Attention-like weighting (simulate GAT by weighting neighbors)
    # We'll use multiple parallel dense layers
    attention_heads = []
    for _ in range(4):  # 4 attention heads
        h = layers.Dense(32, activation='relu')(x)
        attention_heads.append(h)
    
    # Concatenate heads
    if len(attention_heads) > 1:
        x = layers.Concatenate()(attention_heads)
    else:
        x = attention_heads[0]
    
    x = layers.Dense(32, activation='relu')(x)
    embeddings = layers.Dense(embedding_dim, activation='relu', name='embedding')(x)
    
    # Reconstruction
    outputs = layers.Dense(features.shape[1], activation='sigmoid')(embeddings)
    
    model = keras.Model(inputs=inputs, outputs=outputs)
    model.compile(optimizer=Adam(learning_rate=0.01), loss='mse')
    
    # Add adjacency information as a regularizer (simple approach)
    # We'll train with original features but use adjacency for neighbor influence
    model.fit(features, features, epochs=epochs, batch_size=min(32, n),
              verbose=0, shuffle=True)
    
    # Get embeddings
    embedding_model = keras.Model(inputs=inputs, outputs=embeddings)
    node_embeddings = embedding_model.predict(features, verbose=0)
    
    # Apply adjacency smoothing (neighbors should have similar embeddings)
    A_norm = A / (np.sum(A, axis=1, keepdims=True) + 1e-8)
    smoothed_embeddings = np.dot(A_norm, node_embeddings)
    final_embeddings = 0.5 * node_embeddings + 0.5 * smoothed_embeddings
    
    n_communities = max(2, int(np.sqrt(n)))
    partition = embeddings_to_partition(final_embeddings, n_communities)
    
    return partition

# ============================================================
# 6. GAN MODEL
# ============================================================

def gan_community_detection(G: nx.Graph, epochs=150, latent_dim=32, embedding_dim=16) -> Dict[int, int]:
    """Generative Adversarial Network for community detection"""
    print("  Running GAN...")
    
    n = G.number_of_nodes()
    features = get_node_features(G)
    feature_dim = features.shape[1]
    
    # Generator
    def build_generator():
        model = keras.Sequential([
            layers.Dense(64, activation='relu', input_dim=latent_dim),
            layers.BatchNormalization(),
            layers.Dense(128, activation='relu'),
            layers.BatchNormalization(),
            layers.Dense(feature_dim, activation='tanh')
        ])
        return model
    
    # Discriminator
    def build_discriminator():
        model = keras.Sequential([
            layers.Dense(128, activation='relu', input_dim=feature_dim),
            layers.Dropout(0.3),
            layers.Dense(64, activation='relu'),
            layers.Dropout(0.3),
            layers.Dense(1, activation='sigmoid')
        ])
        return model
    
    generator = build_generator()
    discriminator = build_discriminator()
    discriminator.compile(optimizer=Adam(learning_rate=0.0002, beta_1=0.5),
                         loss='binary_crossentropy', metrics=['accuracy'])
    
    # Combined GAN
    discriminator.trainable = False
    gan_input = keras.Input(shape=(latent_dim,))
    fake_features = generator(gan_input)
    gan_output = discriminator(fake_features)
    gan = keras.Model(gan_input, gan_output)
    gan.compile(optimizer=Adam(learning_rate=0.0002, beta_1=0.5), loss='binary_crossentropy')
    
    # Training
    batch_size = min(32, n)
    for epoch in range(epochs):
        # Train discriminator
        noise = np.random.normal(0, 1, size=(batch_size, latent_dim))
        generated = generator.predict(noise, verbose=0)
        
        idx = np.random.randint(0, n, batch_size)
        real = features[idx]
        
        X = np.vstack([real, generated])
        y = np.array([1] * batch_size + [0] * batch_size)
        
        d_loss = discriminator.train_on_batch(X, y)
        
        # Train generator
        noise = np.random.normal(0, 1, size=(batch_size, latent_dim))
        g_loss = gan.train_on_batch(noise, np.ones(batch_size))
        
        if (epoch + 1) % 50 == 0:
            print(f"    Epoch {epoch+1}/{epochs}, D Loss: {d_loss[0]:.4f}, G Loss: {g_loss:.4f}")
    
    # Generate embeddings for all nodes
    all_noise = np.random.normal(0, 1, size=(n, latent_dim))
    generated_features = generator.predict(all_noise, verbose=0)
    
    n_communities = max(2, int(np.sqrt(n)))
    partition = embeddings_to_partition(generated_features, n_communities)
    
    return partition

# ============================================================
# 7. AUTOENCODER MODEL
# ============================================================

def autoencoder_community_detection(G: nx.Graph, epochs=150, encoding_dim=16) -> Dict[int, int]:
    """Stacked Autoencoder for community detection"""
    print("  Running Autoencoder...")
    
    n = G.number_of_nodes()
    features = get_node_features(G)
    feature_dim = features.shape[1]
    
    # Build autoencoder
    input_dim = features.shape[1]
    
    # Encoder
    encoder_input = keras.Input(shape=(input_dim,))
    x = layers.Dense(64, activation='relu')(encoder_input)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(32, activation='relu')(x)
    encoded = layers.Dense(encoding_dim, activation='relu', name='encoded')(x)
    
    # Decoder
    x = layers.Dense(32, activation='relu')(encoded)
    x = layers.Dense(64, activation='relu')(x)
    decoded = layers.Dense(feature_dim, activation='sigmoid')(x)
    
    autoencoder = keras.Model(encoder_input, decoded)
    autoencoder.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
    
    # Train
    autoencoder.fit(features, features, epochs=epochs, batch_size=min(32, n),
                   verbose=0, shuffle=True)
    
    # Get encoder for embeddings
    encoder = keras.Model(encoder_input, encoded)
    node_embeddings = encoder.predict(features, verbose=0)
    
    n_communities = max(2, int(np.sqrt(n)))
    partition = embeddings_to_partition(node_embeddings, n_communities)
    
    return partition

# ============================================================
# 8. MAIN TESTING FUNCTION
# ============================================================

def test_deep_learning_methods(dataset_name: str, G: nx.Graph, ground_truth: Dict[int, int]):
    """Test all deep learning methods on a single dataset"""
    
    methods = {
        "GCN (Graph Convolutional Network)": gcn_community_detection,
        "GAT (Graph Attention Network)": gat_community_detection,
        "GAN (Generative Adversarial Network)": gan_community_detection,
        "Autoencoder (Stacked AE)": autoencoder_community_detection,
    }
    
    results = []
    
    print(f"\n{'='*80}")
    print(f"Testing Deep Learning Methods on: {dataset_name}")
    print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")
    if ground_truth:
        print(f"Ground truth communities: {len(set(ground_truth.values()))}")
    print(f"{'='*80}\n")
    
    for name, method in methods.items():
        try:
            start_time = time.time()
            partition = method(G)
            end_time = time.time()
            runtime = end_time - start_time
            
            n_communities = len(set(partition.values()))
            modularity = calculate_modularity(G, partition)
            
            has_gt = ground_truth and len(ground_truth) > 0
            if has_gt:
                nmi = calculate_nmi(partition, ground_truth)
                ari = calculate_ari(partition, ground_truth)
            else:
                nmi = -1
                ari = -1
            
            if has_gt:
                print(f"✓ {name:35s} | Comm: {n_communities:3d} | Mod: {modularity:.4f} | NMI: {nmi:.4f} | ARI: {ari:.4f} | Time: {runtime:.2f}s")
            else:
                print(f"✓ {name:35s} | Comm: {n_communities:3d} | Mod: {modularity:.4f} | NMI: N/A | ARI: N/A | Time: {runtime:.2f}s")
            
            results.append({
                "Algorithm": name,
                "Communities": n_communities,
                "Modularity": round(modularity, 4),
                "NMI": round(nmi, 4) if has_gt else "N/A",
                "ARI": round(ari, 4) if has_gt else "N/A",
                "Runtime (s)": round(runtime, 2)
            })
            
        except Exception as e:
            print(f"✗ {name:35s} | FAILED: {str(e)[:60]}")
            results.append({
                "Algorithm": name,
                "Communities": "ERROR",
                "Modularity": "ERROR",
                "NMI": "ERROR",
                "ARI": "ERROR",
                "Runtime (s)": "ERROR"
            })
    
    return pd.DataFrame(results)

# ============================================================
# 9. MAIN EXECUTION
# ============================================================

def main():
    print("\n" + "="*80)
    print("DEEP LEARNING COMMUNITY DETECTION COMPARISON")
    print("Methods: GCN, GAT, GAN, Autoencoder")
    print("Metrics: Modularity, NMI, ARI")
    print("Master's Research Project - MOUDDEN Hamza")
    print("="*80)
    
    # Define datasets
    datasets = {
        "Karate Club": ("karate.gml", load_karate_club),
        "Dolphins": ("Dolphins.gml", load_dolphins),
        "Football": ("Football.gml", load_football),
        "Polbooks": ("Polbooks.gml", load_polbooks),
        "Primary School": ("PS.gml", load_primary_school),
        "VS13": ("VS13.gml", load_vs13),
        "VS15": ("VS15.gml", load_vs15),
    }
    
    all_results = {}
    
    for name, (filename, loader) in datasets.items():
        filepath = get_data_path(filename)
        if os.path.exists(filepath):
            print(f"\nLoading {name} from {filename}...")
            try:
                G, gt = loader()
                if G.number_of_nodes() == 0:
                    print(f"  Error: Graph has no nodes")
                    continue
                df = test_deep_learning_methods(name, G, gt)
                all_results[name] = df
                safe_name = name.replace(' ', '_').replace('(', '').replace(')', '')
                df.to_csv(f"deep_results_{safe_name}.csv", index=False)
                print(f"\nResults saved to: deep_results_{safe_name}.csv")
            except Exception as e:
                print(f"  Error loading {name}: {e}")
        else:
            print(f"\n⚠ File not found: {filepath}")
    
    # Create summary
    if all_results:
        print("\n" + "="*80)
        print("SUMMARY: Average Performance Across All Datasets")
        print("="*80)
        
        summary = []
        methods_order = ["GCN (Graph Convolutional Network)", "GAT (Graph Attention Network)", 
                        "GAN (Generative Adversarial Network)", "Autoencoder (Stacked AE)"]
        
        for algo in methods_order:
            mod_values = []
            nmi_values = []
            ari_values = []
            time_values = []
            
            for df in all_results.values():
                row = df[df["Algorithm"] == algo]
                if len(row) > 0 and row["Modularity"].iloc[0] != "ERROR":
                    try:
                        mod_values.append(float(row["Modularity"].iloc[0]))
                        if row["NMI"].iloc[0] != "N/A":
                            nmi_values.append(float(row["NMI"].iloc[0]))
                        if row["ARI"].iloc[0] != "N/A":
                            ari_values.append(float(row["ARI"].iloc[0]))
                        time_values.append(float(row["Runtime (s)"].iloc[0]))
                    except:
                        pass
            
            summary.append({
                "Algorithm": algo,
                "Avg Modularity": f"{np.mean(mod_values):.4f}" if mod_values else "N/A",
                "Avg NMI": f"{np.mean(nmi_values):.4f}" if nmi_values else "N/A",
                "Avg ARI": f"{np.mean(ari_values):.4f}" if ari_values else "N/A",
                "Avg Time (s)": f"{np.mean(time_values):.2f}" if time_values else "N/A"
            })
        
        summary_df = pd.DataFrame(summary)
        print(summary_df.to_string(index=False))
        summary_df.to_csv("deep_results_summary.csv", index=False)
        print("\nSummary saved to: deep_results_summary.csv")
    
    print("\n" + "="*80)
    print("Deep Learning execution completed.")
    print("="*80)

if __name__ == "__main__":
    main()
```
