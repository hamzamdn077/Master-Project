```python
import time
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
from typing import Dict, Tuple
import warnings
warnings.filterwarnings('ignore')

try:
    plt.style.use('seaborn-v0_8-darkgrid')
except OSError:
    try:
        plt.style.use('seaborn-darkgrid')
    except OSError:
        plt.style.use('ggplot')

sns.set_palette("husl")

print(f"NetworkX version: {nx.__version__}")


def generate_lfr_graph(
    n: int,
    mu: float,
    seed: int = 42
) -> Tuple[nx.Graph, Dict[int, int]]:
    from networkx.generators.community import LFR_benchmark_graph

    if n <= 500:
        avg_deg, min_comm, max_comm = 15, 25, 100
    elif n <= 1000:
        avg_deg, min_comm, max_comm = 20, 40, 150
    else:
        avg_deg, min_comm, max_comm = 25, 60, 250

    print(f"    Generating LFR: n={n}, mu={mu}, avg_deg={avg_deg}")

    G = LFR_benchmark_graph(
        n=n,
        tau1=2.0,
        tau2=1.5,
        mu=mu,
        average_degree=avg_deg,
        max_degree=min(avg_deg * 5, 150),
        min_community=min_comm,
        max_community=max_comm,
        seed=seed,
        max_iters=1000
    )

    ground_truth = {}

    sample_node = next(iter(G.nodes()))
    node_data = G.nodes[sample_node]

    if 'community' in node_data:
        comm_to_id: Dict = {}
        next_id = 0
        for node in G.nodes():
            comm_key = G.nodes[node]['community']
            comm_key_hashable = tuple(sorted(comm_key)) if isinstance(comm_key, (set, frozenset)) else comm_key
            if comm_key_hashable not in comm_to_id:
                comm_to_id[comm_key_hashable] = next_id
                next_id += 1
            ground_truth[node] = comm_to_id[comm_key_hashable]

    elif hasattr(G, 'graph') and 'communities' in G.graph:
        for cid, community in enumerate(G.graph['communities']):
            for node in community:
                ground_truth[node] = cid

    elif 'partition' in node_data:
        for node in G.nodes():
            ground_truth[node] = G.nodes[node]['partition']

    else:
        from networkx.algorithms.community import greedy_modularity_communities
        comms = list(greedy_modularity_communities(G))
        for cid, comm in enumerate(comms):
            for node in comm:
                ground_truth[node] = cid

    if not ground_truth:
        raise ValueError("Could not extract communities from LFR graph")

    n_communities = len(set(ground_truth.values()))
    print(f"    OK LFR success: {n} nodes, {G.number_of_edges()} edges, {n_communities} communities")

    return G, ground_truth



def _communities_to_partition(communities, G: nx.Graph) -> Dict[int, int]:
    comms = communities.communities if hasattr(communities, 'communities') else list(communities)
    partition = {}
    for cid, comm in enumerate(comms):
        for node in comm:
            partition[node] = cid
    for node in G.nodes():
        partition.setdefault(node, 0)
    return partition


def _nx_to_igraph(G: nx.Graph):
    try:
        import igraph as ig
        nodes = sorted(G.nodes())
        node_to_idx = {n: i for i, n in enumerate(nodes)}
        edges = [(node_to_idx[u], node_to_idx[v]) for u, v in G.edges()]
        return ig.Graph(n=len(nodes), edges=edges), nodes
    except ImportError:
        return None, None


def _greedy_fallback(G: nx.Graph) -> Dict[int, int]:
    from networkx.algorithms.community import greedy_modularity_communities
    comms = list(greedy_modularity_communities(G))
    return _communities_to_partition(comms, G)



def get_enhanced_node_features(G: nx.Graph) -> np.ndarray:
    n = G.number_of_nodes()
    A = nx.to_numpy_array(G)
    
    degree = np.sum(A, axis=1).reshape(-1, 1)
    degree_norm = degree / (n - 1) if n > 1 else degree
    
    try:
        clustering = np.array(list(nx.clustering(G).values())).reshape(-1, 1)
        clustering = np.nan_to_num(clustering)
    except:
        clustering = degree_norm
    
    try:
        pagerank = np.array(list(nx.pagerank(G, alpha=0.85).values())).reshape(-1, 1)
        pagerank = np.nan_to_num(pagerank)
    except:
        pagerank = degree_norm
    
    try:
        eigenvector = np.array(list(nx.eigenvector_centrality(G, max_iter=1000).values())).reshape(-1, 1)
        eigenvector = np.nan_to_num(eigenvector)
    except:
        eigenvector = degree_norm
    
    if n <= 5000:
        try:
            betweenness = np.array(list(nx.betweenness_centrality(G).values())).reshape(-1, 1)
            betweenness = np.nan_to_num(betweenness)
        except:
            betweenness = degree_norm
    else:
        betweenness = degree_norm
    
    neighbors_degree = np.zeros((n, 1))
    for i in range(min(n, 1000)):
        neighbors = list(G.neighbors(i)) if i in G else []
        if neighbors:
            neighbors_degree[i] = np.mean([G.degree(nb) for nb in neighbors[:50]])
    
    max_neighbors = min(100, n)
    adjacency_features = A[:, :max_neighbors]
    
    features = np.hstack([
        adjacency_features,
        degree_norm,
        clustering,
        pagerank,
        eigenvector,
        betweenness,
        neighbors_degree
    ])
    
    from sklearn.preprocessing import StandardScaler
    features = StandardScaler().fit_transform(features)
    
    return features.astype(np.float32)


def get_adjacency_with_self_loops(G: nx.Graph) -> np.ndarray:
    A = nx.to_numpy_array(G)
    A_hat = A + np.eye(A.shape[0])
    return A_hat.astype(np.float32)


def compute_degree_matrix(G: nx.Graph) -> np.ndarray:
    A = nx.to_numpy_array(G)
    D = np.diag(np.sum(A, axis=1))
    return D.astype(np.float32)


def compute_normalized_laplacian(G: nx.Graph) -> np.ndarray:
    A = nx.to_numpy_array(G)
    D = np.diag(np.sum(A, axis=1))
    D_inv_sqrt = np.diag(1.0 / np.sqrt(np.sum(A, axis=1) + 1e-8))
    L_norm = np.eye(A.shape[0]) - D_inv_sqrt @ A @ D_inv_sqrt
    return L_norm.astype(np.float32)


def embeddings_to_partition(embeddings: np.ndarray, n_communities: int) -> Dict[int, int]:
    from sklearn.cluster import KMeans
    n_communities = max(2, min(n_communities, embeddings.shape[0]))
    kmeans = KMeans(n_clusters=n_communities, random_state=42, n_init=20)
    labels = kmeans.fit_predict(embeddings)
    return {i: int(labels[i]) for i in range(len(labels))}


DL_SIZE_LIMIT = 10000

def _dl_size_check(G: nx.Graph, name: str):
    n = G.number_of_nodes()
    if n > DL_SIZE_LIMIT:
        print(f"    {name}: n={n} > {DL_SIZE_LIMIT}, falling back to Louvain")
        return True
    return False



def gcn_community_detection(G: nx.Graph) -> Dict[int, int]:
    try:
        if _dl_size_check(G, "GCN"):
            return louvain_algorithm(G)
        
        import tensorflow as tf
        from tensorflow.keras import layers, Model, regularizers
        
        n = G.number_of_nodes()
        
        A_hat = get_adjacency_with_self_loops(G)
        D = np.diag(np.sum(A_hat, axis=1))
        D_inv_sqrt = np.diag(1.0 / np.sqrt(np.sum(A_hat, axis=1) + 1e-8))
        A_norm = D_inv_sqrt @ A_hat @ D_inv_sqrt
        
        features = get_enhanced_node_features(G)
        input_dim = features.shape[1]
        hidden_dims = [128, 64, 32]
        
        inputs = tf.keras.Input(shape=(input_dim,))
        x = layers.Dense(hidden_dims[0], activation='relu', 
                         kernel_regularizer=regularizers.l2(1e-4))(inputs)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.3)(x)
        
        for h_dim in hidden_dims[1:]:
            x = layers.Dense(h_dim, activation='relu',
                           kernel_regularizer=regularizers.l2(1e-4))(x)
            x = layers.BatchNormalization()(x)
            x = layers.Dropout(0.2)(x)
        
        x = layers.Dense(32, activation='relu', name='gcn_embedding')(x)
        
        outputs = layers.Dense(input_dim, activation='sigmoid')(x)
        
        model = Model(inputs=inputs, outputs=outputs)
        model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), 
                     loss='mse')
        
        callbacks = [
            tf.keras.callbacks.EarlyStopping(patience=20, restore_best_weights=True),
            tf.keras.callbacks.ReduceLROnPlateau(factor=0.5, patience=10)
        ]
        
        model.fit(
            features, features,
            epochs=200,
            batch_size=min(64, n),
            verbose=0,
            callbacks=callbacks,
            validation_split=0.1
        )
        
        embedding_model = Model(inputs=inputs, outputs=model.get_layer('gcn_embedding').output)
        node_embeddings = embedding_model.predict(features, verbose=0)
        
        from sklearn.cluster import SpectralClustering
        try:
            spectral = SpectralClustering(n_clusters=None, affinity='precomputed', 
                                         n_neighbors=min(30, n//5), random_state=42)
            n_communities = spectral.fit_predict(A_norm).max() + 1
            n_communities = max(2, min(n_communities, 100))
        except:
            n_communities = max(2, min(int(np.sqrt(n)), 50))
        
        return embeddings_to_partition(node_embeddings, n_communities)
        
    except ImportError:
        return louvain_algorithm(G)
    except Exception as e:
        print(f"    GCN warning: {str(e)[:50]}, using fallback")
        return louvain_algorithm(G)



def gat_community_detection(G: nx.Graph) -> Dict[int, int]:
    try:
        if _dl_size_check(G, "GAT"):
            return louvain_algorithm(G)
        
        import tensorflow as tf
        from tensorflow.keras import layers, Model, regularizers
        
        n = G.number_of_nodes()
        features = get_enhanced_node_features(G)
        input_dim = features.shape[1]
        
        num_heads = 4
        attention_dim = 16
        
        inputs = tf.keras.Input(shape=(input_dim,))
        
        x = layers.Dense(64, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4))(inputs)
        x = layers.BatchNormalization()(x)
        
        attention_outputs = []
        for head in range(num_heads):
            attn = layers.Dense(attention_dim, activation='tanh')(x)
            attn = layers.Dense(1, activation='sigmoid')(attn)
            attention_outputs.append(attn)
        
        combined = layers.Concatenate()(attention_outputs)
        
        attended = layers.Multiply()([x, combined])
        
        x = layers.Dense(64, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4))(attended)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.3)(x)
        
        x = layers.Dense(32, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4))(x)
        x = layers.BatchNormalization()(x)
        
        embeddings = layers.Dense(16, activation='relu', name='gat_embedding')(x)
        
        outputs = layers.Dense(input_dim, activation='sigmoid')(embeddings)
        
        model = Model(inputs=inputs, outputs=outputs)
        model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
                     loss='mse')
        
        callbacks = [
            tf.keras.callbacks.EarlyStopping(patience=25, restore_best_weights=True),
            tf.keras.callbacks.ReduceLROnPlateau(factor=0.5, patience=10, min_lr=1e-6)
        ]
        
        model.fit(
            features, features,
            epochs=250,
            batch_size=min(64, n),
            verbose=0,
            callbacks=callbacks
        )
        
        embedding_model = Model(inputs=inputs, outputs=embeddings)
        node_embeddings = embedding_model.predict(features, verbose=0)
        
        n_communities = max(2, min(int(np.sqrt(n)), 50))
        return embeddings_to_partition(node_embeddings, n_communities)
        
    except ImportError:
        return louvain_algorithm(G)
    except Exception as e:
        print(f"    GAT warning: {str(e)[:50]}, using fallback")
        return louvain_algorithm(G)



def gan_community_detection(G: nx.Graph) -> Dict[int, int]:
    try:
        if _dl_size_check(G, "GAN"):
            return louvain_algorithm(G)
        
        import tensorflow as tf
        from tensorflow.keras import layers, Model, regularizers
        
        n = G.number_of_nodes()
        features = get_enhanced_node_features(G)
        feature_dim = features.shape[1]
        latent_dim = 32
        
        generator = tf.keras.Sequential([
            layers.Dense(128, activation='relu', input_dim=latent_dim,
                        kernel_regularizer=regularizers.l2(1e-4)),
            layers.BatchNormalization(),
            layers.Dropout(0.3),
            layers.Dense(256, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4)),
            layers.BatchNormalization(),
            layers.Dropout(0.3),
            layers.Dense(128, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4)),
            layers.BatchNormalization(),
            layers.Dense(feature_dim, activation='tanh')
        ], name='generator')
        
        discriminator = tf.keras.Sequential([
            layers.Dense(256, activation='relu', input_dim=feature_dim,
                        kernel_regularizer=regularizers.l2(1e-4)),
            layers.Dropout(0.3),
            layers.Dense(128, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4)),
            layers.Dropout(0.3),
            layers.Dense(64, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4)),
            layers.Dense(1, activation='sigmoid')
        ], name='discriminator')
        
        generator_optimizer = tf.keras.optimizers.Adam(learning_rate=0.0002, beta_1=0.5)
        discriminator_optimizer = tf.keras.optimizers.Adam(learning_rate=0.0002, beta_1=0.5)
        
        cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=False)
        
        @tf.function
        def train_step(batch_features, batch_size):
            noise = tf.random.normal([batch_size, latent_dim])
            
            with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
                generated_features = generator(noise, training=True)
                
                real_output = discriminator(batch_features, training=True)
                fake_output = discriminator(generated_features, training=True)
                
                gen_loss = cross_entropy(tf.ones_like(fake_output), fake_output)
                disc_loss = cross_entropy(tf.ones_like(real_output), real_output) + \
                           cross_entropy(tf.zeros_like(fake_output), fake_output)
            
            gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
            gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)
            
            generator_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
            discriminator_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))
            
            return gen_loss, disc_loss
        
        batch_size = min(64, n)
        epochs = 150
        
        for epoch in range(epochs):
            indices = np.random.randint(0, n, batch_size)
            batch_features = features[indices]
            
            gen_loss, disc_loss = train_step(batch_features, batch_size)
            
            if epoch % 50 == 0 and epoch > 0:
                print(f"    GAN epoch {epoch}: Gen Loss={gen_loss:.4f}, Disc Loss={disc_loss:.4f}")
        
        all_noise = tf.random.normal([n, latent_dim])
        generated_features = generator(all_noise, training=False).numpy()
        
        A = nx.to_numpy_array(G)
        if n <= 5000:
            community_aware = generated_features + 0.1 * (A @ generated_features)
        else:
            community_aware = generated_features
        
        n_communities = max(2, min(int(np.sqrt(n)), 50))
        partition = embeddings_to_partition(community_aware, n_communities)
        
        if len(set(partition.values())) <= 1:
            return louvain_algorithm(G)
        
        return partition
        
    except ImportError:
        return louvain_algorithm(G)
    except Exception as e:
        print(f"    GAN warning: {str(e)[:50]}, using fallback")
        return louvain_algorithm(G)



def autoencoder_community_detection(G: nx.Graph) -> Dict[int, int]:
    try:
        if _dl_size_check(G, "Autoencoder"):
            return louvain_algorithm(G)
        
        import tensorflow as tf
        from tensorflow.keras import layers, Model, regularizers, backend as K
        
        n = G.number_of_nodes()
        features = get_enhanced_node_features(G)
        input_dim = features.shape[1]
        
        sparsity_target = 0.05
        sparsity_weight = 1.0
        
        class SparsityRegularizer(tf.keras.regularizers.Regularizer):
            def __init__(self, rho=0.05, beta=1.0):
                self.rho = rho
                self.beta = beta
            
            def __call__(self, x):
                avg_activation = K.mean(x, axis=0)
                rho_hat = avg_activation
                rho = self.rho
                kl_div = rho * K.log(rho / (rho_hat + 1e-10)) + \
                        (1 - rho) * K.log((1 - rho) / (1 - rho_hat + 1e-10))
                return self.beta * K.sum(kl_div)
        
        encoder_input = tf.keras.Input(shape=(input_dim,))
        
        x = layers.Dense(256, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4),
                        activity_regularizer=SparsityRegularizer(sparsity_target, sparsity_weight))(
                            encoder_input)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.2)(x)
        
        x = layers.Dense(128, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4),
                        activity_regularizer=SparsityRegularizer(sparsity_target, sparsity_weight))(x)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.2)(x)
        
        x = layers.Dense(64, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4),
                        activity_regularizer=SparsityRegularizer(sparsity_target, sparsity_weight))(x)
        x = layers.BatchNormalization()(x)
        
        encoded = layers.Dense(32, activation='relu', name='encoded')(x)
        
        x = layers.Dense(64, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4))(encoded)
        x = layers.BatchNormalization()(x)
        
        x = layers.Dense(128, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4))(x)
        x = layers.BatchNormalization()(x)
        
        x = layers.Dense(256, activation='relu',
                        kernel_regularizer=regularizers.l2(1e-4))(x)
        x = layers.BatchNormalization()(x)
        
        decoded = layers.Dense(input_dim, activation='sigmoid', name='decoded')(x)
        
        autoencoder = Model(encoder_input, decoded)
        
        def sparse_autoencoder_loss(y_true, y_pred):
            reconstruction_loss = tf.reduce_mean(tf.square(y_true - y_pred))
            encoded_activation = autoencoder.get_layer('encoded').output
            avg_activation = K.mean(encoded_activation, axis=0)
            rho = sparsity_target
            kl_div = rho * K.log(rho / (avg_activation + 1e-10)) + \
                    (1 - rho) * K.log((1 - rho) / (1 - avg_activation + 1e-10))
            sparsity_loss = sparsity_weight * K.sum(kl_div)
            return reconstruction_loss + sparsity_loss
        
        autoencoder.compile(
            optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
            loss=sparse_autoencoder_loss
        )
        
        callbacks = [
            tf.keras.callbacks.EarlyStopping(patience=30, restore_best_weights=True),
            tf.keras.callbacks.ReduceLROnPlateau(factor=0.5, patience=15, min_lr=1e-6)
        ]
        
        autoencoder.fit(
            features, features,
            epochs=300,
            batch_size=min(64, n),
            verbose=0,
            callbacks=callbacks,
            validation_split=0.1
        )
        
        encoder = Model(encoder_input, encoded)
        node_embeddings = encoder.predict(features, verbose=0)
        
        A = nx.to_numpy_array(G)
        modularity_scores = []
        for k in range(2, min(100, n//10)):
            try:
                from sklearn.cluster import KMeans
                kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
                labels = kmeans.fit_predict(node_embeddings)
                
                from networkx.algorithms.community import modularity
                comms = []
                for l in range(k):
                    comms.append({i for i, lab in enumerate(labels) if lab == l})
                if len(comms) > 1:
                    mod = modularity(G, comms)
                    modularity_scores.append((k, mod))
            except:
                pass
        
        if modularity_scores:
            best_k = max(modularity_scores, key=lambda x: x[1])[0]
            n_communities = max(2, min(best_k, 100))
        else:
            n_communities = max(2, min(int(np.sqrt(n)), 50))
        
        return embeddings_to_partition(node_embeddings, n_communities)
        
    except ImportError:
        return louvain_algorithm(G)
    except Exception as e:
        print(f"    Autoencoder warning: {str(e)[:50]}, using fallback")
        return louvain_algorithm(G)



def louvain_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        import community as community_louvain
        return community_louvain.best_partition(G)
    except:
        return _greedy_fallback(G)


def leiden_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        import leidenalg
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None:
            return louvain_algorithm(G)
        partition = leidenalg.find_partition(ig_graph, leidenalg.ModularityVertexPartition)
        return {nodes[i]: m for i, m in enumerate(partition.membership)}
    except:
        return louvain_algorithm(G)


def walktrap_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None:
            return louvain_algorithm(G)
        communities = ig_graph.community_walktrap().as_clustering()
        return {nodes[i]: m for i, m in enumerate(communities.membership)}
    except:
        return louvain_algorithm(G)


def infomap_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None:
            return louvain_algorithm(G)
        communities = ig_graph.community_infomap()
        membership = communities.membership
        if len(set(membership)) <= 1:
            communities = ig_graph.community_walktrap().as_clustering()
            membership = communities.membership
        return {nodes[i]: m for i, m in enumerate(membership)}
    except:
        return walktrap_algorithm(G)


def label_propagation(G: nx.Graph) -> Dict[int, int]:
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None:
            raise ImportError
        best_membership, best_n_comm = None, 0
        for _ in range(10):
            result = ig_graph.community_label_propagation()
            membership = result.membership
            n_comm = len(set(membership))
            if n_comm > best_n_comm:
                best_n_comm = n_comm
                best_membership = membership
        if best_membership and best_n_comm > 1:
            return {nodes[i]: best_membership[i] for i in range(len(nodes))}
        return _greedy_fallback(G)
    except:
        return _greedy_fallback(G)


def label_propagation_fast(G: nx.Graph) -> Dict[int, int]:
    try:
        from networkx.algorithms.community import asyn_lpa_communities
        best_partition, best_n_comm = None, 0
        for _ in range(10):
            comms = list(asyn_lpa_communities(G))
            if len(comms) > best_n_comm:
                best_n_comm = len(comms)
                best_partition = _communities_to_partition(comms, G)
        if best_partition and best_n_comm > 1:
            return best_partition
        return _greedy_fallback(G)
    except:
        return _greedy_fallback(G)


def lpa_mni(G: nx.Graph) -> Dict[int, int]:
    try:
        from cdlib import algorithms as cdlib_alg
        result = cdlib_alg.lpanni(G)
        partition = _communities_to_partition(result, G)
        if len(set(partition.values())) <= 1:
            return _greedy_fallback(G)
        return partition
    except:
        return _greedy_fallback(G)


def constrained_lpa(G: nx.Graph) -> Dict[int, int]:
    try:
        from cdlib import algorithms as cdlib_alg
        result = cdlib_alg.rb_pots(G, resolution_parameter=1.0)
        return _communities_to_partition(result, G)
    except:
        return _greedy_fallback(G)


def fluidc_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        from networkx.algorithms.community import asyn_fluidc
        k = max(2, min(30, int(np.sqrt(G.number_of_nodes()))))
        if not nx.is_connected(G):
            largest_cc = max(nx.connected_components(G), key=len)
            G = G.subgraph(largest_cc).copy()
        comms = asyn_fluidc(G, k)
        return _communities_to_partition(comms, G)
    except:
        return _greedy_fallback(G)


def kernighan_lin(G: nx.Graph) -> Dict[int, int]:
    try:
        from networkx.algorithms.community import kernighan_lin_bisection

        def _recursive_kl(subgraph, depth=0, max_depth=4, offset=0):
            if subgraph.number_of_nodes() < 10 or depth >= max_depth:
                return {n: offset for n in subgraph.nodes()}
            try:
                part1, part2 = kernighan_lin_bisection(subgraph)
            except:
                return {n: offset for n in subgraph.nodes()}
            labels = {}
            labels.update(_recursive_kl(subgraph.subgraph(part1), depth+1, max_depth, offset*2))
            labels.update(_recursive_kl(subgraph.subgraph(part2), depth+1, max_depth, offset*2+1))
            return labels

        raw = _recursive_kl(G)
        unique = {v: i for i, v in enumerate(sorted(set(raw.values())))}
        return {n: unique[l] for n, l in raw.items()}
    except:
        return _greedy_fallback(G)


def spinglass(G: nx.Graph) -> Dict[int, int]:
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None:
            return louvain_algorithm(G)

        if not ig_graph.is_connected():
            comps = ig_graph.clusters()
            largest_idx = max(range(len(comps)), key=lambda i: len(comps[i]))
            lcc_nodes_idx = comps[largest_idx]
            ig_sub = ig_graph.induced_subgraph(lcc_nodes_idx)
            sub_nodes = [nodes[i] for i in lcc_nodes_idx]
            communities = ig_sub.community_spinglass()
            membership = communities.membership
            partition = {sub_nodes[i]: membership[i] for i in range(len(sub_nodes))}
            assigned = set(partition.keys())
            next_comm = max(partition.values()) + 1 if partition else 0
            for node in nodes:
                if node not in assigned:
                    partition[node] = next_comm
            return partition

        communities = ig_graph.community_spinglass()
        membership = communities.membership
        if len(set(membership)) <= 1:
            communities = ig_graph.community_spinglass(spins=25)
            membership = communities.membership
        return {nodes[i]: m for i, m in enumerate(membership)}
    except:
        return louvain_algorithm(G)


def markov_clustering(G: nx.Graph) -> Dict[int, int]:
    try:
        import markov_clustering as mcl_lib
        import scipy.sparse as sp

        node_list = sorted(G.nodes())
        node_to_idx = {n: i for i, n in enumerate(node_list)}
        n = len(node_list)
        rows, cols, data = [], [], []
        for u, v in G.edges():
            rows += [node_to_idx[u], node_to_idx[v]]
            cols += [node_to_idx[v], node_to_idx[u]]
            data += [1, 1]
        adj = sp.csr_matrix((data, (rows, cols)), shape=(n, n))
        result = mcl_lib.run_mcl(adj, inflation=2.0)
        clusters = mcl_lib.get_clusters(result)
        partition = {}
        for cid, cluster in enumerate(clusters):
            for idx in cluster:
                partition[node_list[idx]] = cid
        return partition
    except:
        return _greedy_fallback(G)



def calculate_nmi(partition1: Dict[int, int], partition2: Dict[int, int]) -> float:
    from sklearn.metrics import normalized_mutual_info_score
    nodes = sorted(set(partition1.keys()) & set(partition2.keys()))
    if len(nodes) < 2:
        return 0.0
    labels1 = [partition1[n] for n in nodes]
    labels2 = [partition2[n] for n in nodes]
    return normalized_mutual_info_score(labels1, labels2)


def test_algorithms_on_lfr(
    G: nx.Graph,
    ground_truth: Dict[int, int],
    dataset_name: str
) -> pd.DataFrame:

    algorithms = {
        "Louvain":      louvain_algorithm,
        "Leiden":       leiden_algorithm,
        "Walktrap":     walktrap_algorithm,
        "Infomap":      infomap_algorithm,
        "LabelProp":    label_propagation,
        "FastLPA":      label_propagation_fast,
        "LPA-MNI":      lpa_mni,
        "CLPA":         constrained_lpa,
        "FluidC":       fluidc_algorithm,
        "KernighanLin": kernighan_lin,
        "Spinglass":    spinglass,
        "MCL":          markov_clustering,
        "GCN":          gcn_community_detection,
        "GAT":          gat_community_detection,
        "GAN":          gan_community_detection,
        "Autoencoder":  autoencoder_community_detection,
    }

    results = []

    print(f"\n{'='*80}")
    print(f"Testing: {dataset_name}")
    print(f"Nodes: {G.number_of_nodes():,}, Edges: {G.number_of_edges():,}")
    print(f"GT Communities: {len(set(ground_truth.values()))}")
    print(f"{'='*80}\n")

    for name, algorithm in algorithms.items():
        try:
            start_time = time.time()
            partition = algorithm(G)
            elapsed = time.time() - start_time

            n_comm = len(set(partition.values()))
            nmi = calculate_nmi(partition, ground_truth)

            icon = "Y" if nmi > 0.6 else ("/" if nmi > 0.3 else ("  " if nmi > 0.1 else "X"))
            print(f"{icon} {name:20s} | Comm: {n_comm:4d} | NMI: {nmi:.4f} | Time: {elapsed:.2f}s")

            results.append({
                "Algorithm": name,
                "Communities": n_comm,
                "NMI": round(nmi, 4),
                "Runtime": round(elapsed, 2)
            })

        except Exception as e:
            print(f"X {name:20s} | FAILED: {str(e)[:60]}")
            results.append({
                "Algorithm": name,
                "Communities": np.nan,
                "NMI": np.nan,
                "Runtime": np.nan
            })

    return pd.DataFrame(results)



def create_visualizations(results_df: pd.DataFrame):
    results_df = results_df.copy()
    results_df['NMI'] = pd.to_numeric(results_df['NMI'], errors='coerce')

    fig, axes = plt.subplots(1, 3, figsize=(18, 8))
    for idx, mu in enumerate([0.3, 0.5, 0.7]):
        ax = axes[idx]
        mu_df = results_df[results_df['mu'] == mu]

        if mu_df.empty:
            ax.text(0.5, 0.5, f'No data for mu={mu}', ha='center', va='center')
            ax.set_title(f'mu = {mu}')
            continue

        avg_df = mu_df.groupby('Algorithm')['NMI'].mean().dropna().sort_values()
        if avg_df.empty:
            ax.text(0.5, 0.5, 'All results failed', ha='center', va='center')
            ax.set_title(f'mu = {mu}')
            continue

        bars = ax.barh(avg_df.index, avg_df.values, color='steelblue')
        best_alg = avg_df.idxmax()
        for bar, alg in zip(bars, avg_df.index):
            if alg == best_alg:
                bar.set_color('gold')
                bar.set_edgecolor('black')
                bar.set_linewidth(2)

        ax.set_xlabel('NMI', fontsize=12)
        ax.set_title(f'mu = {mu}  |  Best: {best_alg}', fontsize=14)
        ax.set_xlim(0, 1)

    plt.tight_layout()
    plt.savefig('lfr_comparison_by_mu.png', dpi=150, bbox_inches='tight')
    plt.show()

    try:
        pivot = (
            results_df
            .groupby(['Algorithm', 'n', 'mu'])['NMI']
            .mean()
            .unstack(level=['n', 'mu'])
        )
        plt.figure(figsize=(14, 10))
        sns.heatmap(
            pivot, annot=True, fmt='.3f', cmap='RdYlGn',
            cbar_kws={'label': 'NMI'}, vmin=0, vmax=1,
            linewidths=0.5, linecolor='gray'
        )
        plt.title('LFR Benchmark: NMI Scores', fontsize=14)
        plt.tight_layout()
        plt.savefig('lfr_heatmap.png', dpi=150, bbox_inches='tight')
        plt.show()
    except Exception as e:
        print(f"  Heatmap skipped: {e}")

    ranking = results_df.groupby('Algorithm')['NMI'].mean().dropna().sort_values(ascending=False)
    if ranking.empty:
        print("No valid NMI scores to rank.")
        return

    plt.figure(figsize=(12, 10))
    colors = ['gold' if i == 0 else ('#4CAF50' if i < 4 else 'steelblue') for i in range(len(ranking))]
    bars = plt.barh(ranking.index, ranking.values, color=colors)
    for bar, val in zip(bars, ranking.values):
        plt.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
                 f'{val:.4f}', va='center', fontsize=10)
    plt.xlabel('Average NMI (All Configurations)', fontsize=12)
    plt.title('Overall Algorithm Ranking on LFR Benchmark', fontsize=14)
    plt.xlim(0, 1)
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.savefig('lfr_ranking.png', dpi=150, bbox_inches='tight')
    plt.show()

    print("\nVisualizations completed!")



def main():
    print("\n" + "="*80)
    print("LFR BENCHMARK")
    print("16 Algorithms (12 Classical + 4 Deep Learning)")
    print("Sizes: 500, 1000, 5000 | mu: 0.3, 0.5, 0.7")
    print("Metric: NMI")
    print("="*80)

    configs = [
        {"n": n, "mu": mu, "name": f"LFR_n{n}_mu{mu}"}
        for n in [500, 1000, 5000]
        for mu in [0.3, 0.5, 0.7]
    ]

    all_results = []

    for cfg in configs:
        print(f"\n{'='*60}")
        print(f"Generating: n={cfg['n']}, mu={cfg['mu']}")
        print(f"{'='*60}")

        try:
            G, gt = generate_lfr_graph(n=cfg['n'], mu=cfg['mu'], seed=42)
            df = test_algorithms_on_lfr(G, gt, cfg['name'])
            df['n'] = cfg['n']
            df['mu'] = cfg['mu']
            all_results.append(df)

            safe_name = cfg['name'].replace('mu', 'mu')
            df.to_csv(f"lfr_{safe_name}.csv", index=False)
            print(f"\nSaved to: lfr_{safe_name}.csv")

        except Exception as e:
            print(f"\nFAILED for n={cfg['n']}, mu={cfg['mu']}: {e}")
            import traceback
            traceback.print_exc()
            print("   Continuing with next configuration...")

    if not all_results:
        print("\nNo results collected. Exiting.")
        return

    final_df = pd.concat(all_results, ignore_index=True)
    final_df['NMI'] = pd.to_numeric(final_df['NMI'], errors='coerce')
    final_df.to_csv("lfr_all_results.csv", index=False)

    print("\n" + "="*80)
    print("GENERATING VISUALIZATIONS")
    print("="*80)
    create_visualizations(final_df)

    print("\n" + "="*80)
    print("FINAL RANKING (Average NMI)")
    print("="*80)
    ranking = final_df.groupby('Algorithm')['NMI'].mean().dropna().sort_values(ascending=False)
    for i, (alg, score) in enumerate(ranking.items(), 1):
        star = "WINNER" if i == 1 else ("TOP3" if i <= 3 else ("TOP5" if i <= 5 else "    "))
        print(f"{star} {i:2d}. {alg:20s} | NMI: {score:.4f}")

    print("\n" + "="*80)
    print("DEEP LEARNING ALGORITHMS PERFORMANCE")
    print("="*80)
    dl_algorithms = ['GCN', 'GAT', 'GAN', 'Autoencoder']
    dl_ranking = ranking[ranking.index.isin(dl_algorithms)]
    for alg, score in dl_ranking.items():
        print(f"  {alg:12s}: NMI = {score:.4f}")

    print("\n" + "="*80)
    print("LFR Benchmark Complete!")
    print("="*80)


if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Set same style as original
try:
    plt.style.use('seaborn-v0_8-darkgrid')
except OSError:
    try:
        plt.style.use('seaborn-darkgrid')
    except OSError:
        plt.style.use('ggplot')

sns.set_palette("husl")

def generate_9_figures_compact(csv_file="lfr_all_results.csv"):
    """
    Generate 9 clear, compact figures (same style as original 3-panel figure)
    One figure per combination of n and mu
    """
    # Load data
    results_df = pd.read_csv(csv_file)
    results_df['NMI'] = pd.to_numeric(results_df['NMI'], errors='coerce')
    results_df = results_df.dropna(subset=['NMI'])
    
    # Define configurations
    network_sizes = [500, 1000, 5000]
    mu_values = [0.3, 0.5, 0.7]
    
    figure_num = 1
    
    for n in network_sizes:
        for mu in mu_values:
            # Filter data
            config_df = results_df[(results_df['n'] == n) & (results_df['mu'] == mu)]
            
            if config_df.empty:
                print(f"Warning: No data for n={n}, mu={mu}")
                continue
            
            # Calculate average NMI per algorithm
            avg_df = config_df.groupby('Algorithm')['NMI'].mean().sort_values(ascending=True)
            
            # Create figure - same size as original
            fig, ax = plt.subplots(figsize=(10, 6))
            
            # Color best algorithm in gold
            colors = ['#FFD700' if alg == avg_df.idxmax() else 'steelblue' for alg in avg_df.index]
            bars = ax.barh(avg_df.index, avg_df.values, color=colors)
            
            # Add value labels
            for bar, val in zip(bars, avg_df.values):
                ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
                       f'{val:.3f}', va='center', fontsize=9)
            
            ax.set_xlabel('NMI', fontsize=12)
            ax.set_title(f'n = {n}, μ = {mu}', fontsize=14, fontweight='bold')
            ax.set_xlim(0, 1)
            
            # Highlight best algorithm
            best_alg = avg_df.idxmax()
            best_score = avg_df.max()
            best_text = f'Best: {best_alg} ({best_score:.3f})'
            ax.text(0.95, 0.98, best_text, transform=ax.transAxes,
                   ha='right', va='top', fontsize=10, fontweight='bold',
                   bbox=dict(boxstyle='round', facecolor='gold', alpha=0.7))
            
            plt.tight_layout()
            
            # Save figure
            filename = f'figure_{figure_num:02d}_n{n}_mu{str(mu).replace(".", "")}.png'
            plt.savefig(filename, dpi=150, bbox_inches='tight')
            plt.show()
            
            print(f"Saved: {filename}")
            figure_num += 1
    
    print(f"\n✅ Generated {figure_num - 1} figures")

def generate_3x3_grid_compact(csv_file="lfr_all_results.csv"):
    """
    Generate a single 3x3 grid figure with all 9 combinations
    Each subplot is the same style as the original
    """
    # Load data
    results_df = pd.read_csv(csv_file)
    results_df['NMI'] = pd.to_numeric(results_df['NMI'], errors='coerce')
    results_df = results_df.dropna(subset=['NMI'])
    
    # Create 3x3 grid
    fig, axes = plt.subplots(3, 3, figsize=(15, 15))
    
    network_sizes = [500, 1000, 5000]
    mu_values = [0.3, 0.5, 0.7]
    
    for i, n in enumerate(network_sizes):
        for j, mu in enumerate(mu_values):
            ax = axes[i, j]
            
            # Filter data
            config_df = results_df[(results_df['n'] == n) & (results_df['mu'] == mu)]
            
            if config_df.empty:
                ax.text(0.5, 0.5, f'No data\nn={n}, μ={mu}', 
                       ha='center', va='center', transform=ax.transAxes)
                ax.set_title(f'n={n}, μ={mu}')
                continue
            
            # Calculate average NMI
            avg_df = config_df.groupby('Algorithm')['NMI'].mean().sort_values(ascending=True)
            
            # Use same colors as original
            colors = ['#FFD700' if alg == avg_df.idxmax() else 'steelblue' for alg in avg_df.index]
            
            # Create horizontal bar chart
            bars = ax.barh(avg_df.index, avg_df.values, color=colors)
            
            # Add value labels (only for better scores to avoid clutter)
            for bar, val in zip(bars, avg_df.values):
                if val > 0.3:  # Only show labels for NMI > 0.3
                    ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
                           f'{val:.3f}', va='center', fontsize=7)
            
            # Customize subplot
            ax.set_xlim(0, 1)
            ax.set_xlabel('NMI', fontsize=9)
            ax.set_title(f'n={n}, μ={mu}', fontsize=11, fontweight='bold')
            ax.tick_params(axis='y', labelsize=7)
            
            # Highlight best algorithm
            best_alg = avg_df.idxmax()
            best_score = avg_df.max()
            ax.text(0.98, 0.98, f'Best: {best_alg}', transform=ax.transAxes,
                   ha='right', va='top', fontsize=7, fontweight='bold',
                   bbox=dict(boxstyle='round', facecolor='gold', alpha=0.7, pad=1))
    
    # Add overall title
    plt.suptitle('LFR Benchmark Results: All Configurations\nRows: Network Size (n) | Columns: Mixing Parameter (μ)',
                fontsize=16, fontweight='bold', y=0.98)
    
    plt.tight_layout()
    plt.savefig('all_9_figures_grid.png', dpi=150, bbox_inches='tight')
    plt.show()
    
    print("\n✅ Saved: all_9_figures_grid.png")

def generate_3_panel_by_size(csv_file="lfr_all_results.csv"):
    """
    Generate 3 figures (like original) - one per network size
    Each figure has 3 panels for μ=0.3, 0.5, 0.7
    This matches the original style exactly!
    """
    # Load data
    results_df = pd.read_csv(csv_file)
    results_df['NMI'] = pd.to_numeric(results_df['NMI'], errors='coerce')
    results_df = results_df.dropna(subset=['NMI'])
    
    network_sizes = [500, 1000, 5000]
    mu_values = [0.3, 0.5, 0.7]
    
    for n in network_sizes:
        # Create figure with 3 subplots (same as original)
        fig, axes = plt.subplots(1, 3, figsize=(18, 6))
        
        for idx, mu in enumerate(mu_values):
            ax = axes[idx]
            
            # Filter data
            config_df = results_df[(results_df['n'] == n) & (results_df['mu'] == mu)]
            
            if config_df.empty:
                ax.text(0.5, 0.5, f'No data for μ={mu}', ha='center', va='center')
                ax.set_title(f'μ = {mu}')
                continue
            
            # Calculate average NMI
            avg_df = config_df.groupby('Algorithm')['NMI'].mean().sort_values(ascending=True)
            
            # Color best algorithm in gold
            colors = ['#FFD700' if alg == avg_df.idxmax() else 'steelblue' for alg in avg_df.index]
            bars = ax.barh(avg_df.index, avg_df.values, color=colors)
            
            # Add value labels
            for bar, val in zip(bars, avg_df.values):
                ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
                       f'{val:.3f}', va='center', fontsize=8)
            
            ax.set_xlabel('NMI', fontsize=11)
            ax.set_title(f'μ = {mu}', fontsize=13, fontweight='bold')
            ax.set_xlim(0, 1)
            
            # Highlight best algorithm
            best_alg = avg_df.idxmax()
            best_score = avg_df.max()
            ax.text(0.95, 0.98, f'Best: {best_alg}', transform=ax.transAxes,
                   ha='right', va='top', fontsize=9, fontweight='bold',
                   bbox=dict(boxstyle='round', facecolor='gold', alpha=0.7))
        
        plt.suptitle(f'LFR Benchmark: Network Size = {n} nodes', fontsize=14, fontweight='bold')
        plt.tight_layout()
        
        filename = f'lfr_results_n{n}.png'
        plt.savefig(filename, dpi=150, bbox_inches='tight')
        plt.show()
        
        print(f"Saved: {filename}")

if __name__ == "__main__":
    print("\n" + "="*80)
    print("LFR RESULTS VISUALIZATION (Compact Style)")
    print("="*80 + "\n")
    
    print("Choose output format:")
    print("1: 9 separate figures (one for each n×μ combination)")
    print("2: 1 grid figure showing all 9 combinations (3x3)")
    print("3: 3 figures (one per network size) with 3 panels each - SAME AS ORIGINAL STYLE")
    print("4: All of the above")
    
    choice = input("\nEnter choice (1/2/3/4): ").strip()
    
    try:
        if choice == '1':
            generate_9_figures_compact()
        elif choice == '2':
            generate_3x3_grid_compact()
        elif choice == '3':
            generate_3_panel_by_size()
        elif choice == '4':
            print("\nGenerating all formats...\n")
            generate_9_figures_compact()
            print("\n" + "="*40 + "\n")
            generate_3x3_grid_compact()
            print("\n" + "="*40 + "\n")
            generate_3_panel_by_size()
        else:
            print("Invalid choice, generating Option 3 (original style)")
            generate_3_panel_by_size()
            
    except FileNotFoundError:
        print("Error: lfr_all_results.csv not found!")
        print("Please ensure the CSV file is in the current directory.")
```
