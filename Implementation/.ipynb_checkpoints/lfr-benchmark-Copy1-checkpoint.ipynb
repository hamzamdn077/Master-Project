Benchmarks **9 classical** + **4 deep-learning** community-detection algorithms  
on synthetic LFR graphs with three mixing parameters (μ = 0.3 / 0.5 / 0.7).

| | Algorithms |
|---|---|
| **Classical (9)** | Louvain · Leiden · Walktrap · Infomap · LabelProp · FastLPA · FluidC · MCL · Kernighan–Lin |

### Two-Phase Protocol
| Phase | Graph sizes | Notes |
|---|---|---|
| **Phase 1** | 500 · 1 000 · 5 000 | All 13 algorithms — full benchmark |
| **Phase 2** | 10 000 · 50 000 · 100 000 | Top-N promoted algorithms only |

> Every algorithm has a **per-run wall-clock timeout**; slow algorithms are skipped automatically.

## 2. Training Progress Callback

```python
class TrainingProgress(keras.callbacks.Callback):
    """Prints a compact progress bar every `print_every` epochs."""

    def __init__(self, epochs: int, model_name: str = "", print_every: int = 5):
        super().__init__()
        self.total_epochs = epochs
        self.model_name   = model_name
        self.print_every  = print_every
        self.start_time   = None

    def on_train_begin(self, logs=None):
        self.start_time = time.time()
        print(f"    [{self.model_name}] Training started — {self.total_epochs} epochs",
              flush=True)

    def on_epoch_end(self, epoch, logs=None):
        if (epoch + 1) % self.print_every == 0 or (epoch + 1) == self.total_epochs:
            elapsed  = time.time() - self.start_time
            loss     = logs.get("loss", float("nan"))
            val_loss = logs.get("val_loss", None)
            bar_len  = 20
            filled   = int(bar_len * (epoch + 1) / self.total_epochs)
            bar      = "█" * filled + "░" * (bar_len - filled)
            pct      = 100 * (epoch + 1) / self.total_epochs
            val_str  = f" | val_loss: {val_loss:.4f}" if val_loss is not None else ""
            print(f"    [{self.model_name}] [{bar}] {pct:5.1f}%  "
                  f"epoch {epoch+1:>4}/{self.total_epochs}  "
                  f"loss: {loss:.4f}{val_str}  ({elapsed:.1f}s elapsed)", flush=True)

    def on_train_end(self, logs=None):
        print(f"    [{self.model_name}] Training complete in "
              f"{time.time()-self.start_time:.1f}s", flush=True)
```

## 3. LFR Graph Generator

```python
def generate_lfr_graph(n: int, mu: float,
                        seed: int = SEED) -> Tuple[nx.Graph, Dict[int, int]]:
    """
    Generate an LFR benchmark graph and return (G, ground_truth_partition).
    Degree / community-size parameters scale automatically with n.
    """
    from networkx.generators.community import LFR_benchmark_graph

    if n <= 500:
        avg_deg, min_comm, max_comm = 15, 20,  80
    elif n <= 1_000:
        avg_deg, min_comm, max_comm = 20, 30, 120
    elif n <= 5_000:
        avg_deg, min_comm, max_comm = 25, 50, 200
    elif n <= 10_000:
        avg_deg, min_comm, max_comm = 25, 80, 400
    elif n <= 50_000:
        avg_deg, min_comm, max_comm = 20, 100, 600
    else:
        avg_deg, min_comm, max_comm = 15, 150, 800

    print(f"  Generating LFR: n={n:,}, μ={mu}, avg_deg={avg_deg}", end=" ... ", flush=True)
    t0 = time.time()

    # At high mu the LFR sampler struggles to converge — cap iters tighter
    # to avoid burning memory on thousands of failed rewiring passes
    max_iters = 500 if mu >= 0.7 else 2000

    G = LFR_benchmark_graph(
        n=n, tau1=2.0, tau2=1.5, mu=mu,
        average_degree=avg_deg,
        max_degree=min(avg_deg * 5, 200),
        min_community=min_comm,
        max_community=max_comm,
        seed=seed, max_iters=max_iters,
    )

    # Extract ground-truth partition
    ground_truth: Dict[int, int] = {}
    comm_to_id:   Dict           = {}
    nid = 0
    for node in G.nodes():
        raw = G.nodes[node].get("community", node)
        key = tuple(sorted(raw)) if isinstance(raw, (set, frozenset)) else raw
        if key not in comm_to_id:
            comm_to_id[key] = nid; nid += 1
        ground_truth[node] = comm_to_id[key]

    if not ground_truth:
        raise ValueError("Could not extract communities from LFR graph")

    G = G.to_undirected()
    G.remove_edges_from(nx.selfloop_edges(G))

    # ── Strip community frozensets from every node immediately ────────────
    # LFR attaches a frozenset of ALL co-members to each node.
    # At n=50k with avg community size 600 this costs ~800 MB that would
    # otherwise stay live until G is garbage-collected.
    for v in G.nodes():
        G.nodes[v].pop("community", None)

    k_gt = len(set(ground_truth.values()))
    print(f"OK  ({G.number_of_edges():,} edges, {k_gt} communities, {time.time()-t0:.1f}s)")
    return G, ground_truth
```

## 4. Helpers, Evaluation Metrics & Timeout Runner

```python
# ── Partition helpers ─────────────────────────────────────────────────────────

def _communities_to_partition(communities, G: nx.Graph) -> Dict[int, int]:
    comms = (communities.communities
             if hasattr(communities, "communities") else list(communities))
    partition: Dict[int, int] = {}
    for cid, comm in enumerate(comms):
        for node in comm:
            partition[node] = cid
    for node in G.nodes():
        partition.setdefault(node, 0)
    return partition


# Weak-ref cache: build the igraph object once per NX graph object,
# reuse across all algorithm calls within the same benchmark run.
import weakref as _weakref
_igraph_cache: dict = {}   # id(G) → (ig_graph, nodes)

def _nx_to_igraph(G: nx.Graph):
    try:
        import igraph as ig
        gid = id(G)
        if gid in _igraph_cache:
            return _igraph_cache[gid]
        nodes = sorted(G.nodes())
        idx   = {n: i for i, n in enumerate(nodes)}
        edges = [(idx[u], idx[v]) for u, v in G.edges()]
        result = ig.Graph(n=len(nodes), edges=edges), nodes
        _igraph_cache[gid] = result
        return result
    except ImportError:
        return None, None

def _clear_igraph_cache():
    """Call after each benchmark run to release the cached igraph object."""
    _igraph_cache.clear()


def _greedy_fallback(G: nx.Graph) -> Dict[int, int]:
    from networkx.algorithms.community import greedy_modularity_communities
    return _communities_to_partition(
        list(greedy_modularity_communities(G)), G)


# ── Metrics ───────────────────────────────────────────────────────────────────

def calculate_nmi(p1: Dict[int, int], p2: Dict[int, int]) -> float:
    nodes = sorted(set(p1) & set(p2))
    if len(nodes) < 2: return 0.0
    return normalized_mutual_info_score(
        [p1[n] for n in nodes], [p2[n] for n in nodes])



# ── Memory cleanup helper ────────────────────────────────────────────────────

def _clear_keras_session():
    """Destroy all TF graphs accumulated in the session and force a GC pass."""
    keras.backend.clear_session()
    gc.collect()

# ── Wall-clock timeout runner ─────────────────────────────────────────────────

class _TimeoutError(Exception):
    pass

def _sigalrm(signum, frame):
    raise _TimeoutError()

def run_with_timeout(fn, args=(), kwargs=None, timeout_sec: int = 120):
    """
    Run fn(*args, **kwargs) within timeout_sec wall-clock seconds.
    Uses SIGALRM on Linux/Mac; thread-join on Windows.
    Returns (result, elapsed_seconds) or raises _TimeoutError.
    """
    kwargs = kwargs or {}
    if platform.system() != "Windows":
        signal.signal(signal.SIGALRM, _sigalrm)
        signal.alarm(timeout_sec)
        try:
            t0     = time.time()
            result = fn(*args, **kwargs)
            signal.alarm(0)
            return result, time.time() - t0
        except _TimeoutError:
            raise
        finally:
            signal.alarm(0)
    else:
        # Windows: threads cannot be killed, but we can at least:
        # 1. Use a tighter poll loop so we stop waiting as soon as done
        # 2. Abort early if RAM is critically low (prevents OOM stalls)
        result_box, exc_box = [None], [None]
        def _run():
            try:    result_box[0] = fn(*args, **kwargs)
            except Exception as e: exc_box[0] = e
        t0 = time.time()
        t  = threading.Thread(target=_run, daemon=True)
        t.start()
        deadline = t0 + timeout_sec
        while t.is_alive():
            remaining = deadline - time.time()
            if remaining <= 0:
                # Thread still running — abandon it (daemon → dies with kernel)
                raise _TimeoutError(f"Exceeded {timeout_sec}s")
            # Abort early if available RAM drops below 600 MB
            try:
                import psutil
                if psutil.virtual_memory().available < 600 * 1024**2:
                    raise _TimeoutError("RAM critically low — aborting to protect kernel")
            except ImportError:
                pass
            t.join(timeout=min(2.0, remaining))
        elapsed = time.time() - t0
        if exc_box[0]:
            raise exc_box[0]
        return result_box[0], elapsed
```

## 5. Feature Engineering

- **`get_node_features`** — full adjacency row + degree + clustering for n ≤ `DL_DENSE_LIMIT`
- **`get_large_graph_features`** — 5 lightweight structural features for larger graphs
- **`pick_features`** — automatic routing by graph size
- **`embeddings_to_partition`** — uses ground-truth k when available (eliminates the √n bug)


```python
def get_node_features(G: nx.Graph) -> np.ndarray:
    """
    Full feature matrix: adjacency row + normalised degree + clustering coeff.
    Used on small/medium graphs (n ≤ DL_DENSE_LIMIT).
    """
    n      = G.number_of_nodes()
    A      = nx.to_numpy_array(G)
    degree = (np.sum(A, axis=1) / (n - 1 if n > 1 else 1)).reshape(-1, 1)
    clust  = np.array(list(nx.clustering(G).values())).reshape(-1, 1)
    return StandardScaler().fit_transform(
        np.hstack([A, degree, clust])).astype(np.float32)


def get_large_graph_features(G: nx.Graph) -> np.ndarray:
    """
    Lightweight structural features for large graphs (n > DL_DENSE_LIMIT).
    Uses vectorised NetworkX calls (not per-node loops) to avoid O(n²) stalls.
    Features: degree, clustering, triangles, pagerank, betweenness (approx).
    """
    nodes = sorted(G.nodes())
    n     = len(nodes)
    idx   = {v: i for i, v in enumerate(nodes)}

    # degree — O(n)
    deg_dict = dict(G.degree())
    deg  = np.array([deg_dict[v] for v in nodes], dtype=np.float32).reshape(-1, 1)

    # clustering + triangles — single vectorised pass, O(m·d)
    clust_dict = nx.clustering(G)          # returns dict for ALL nodes at once
    tri_dict   = nx.triangles(G)           # same
    clust = np.array([clust_dict[v] for v in nodes], dtype=np.float32).reshape(-1, 1)
    tri   = np.array([tri_dict[v]   for v in nodes], dtype=np.float32).reshape(-1, 1)
    del clust_dict, tri_dict

    # pagerank — cap iterations tighter for large graphs
    try:
        pr   = nx.pagerank(G, max_iter=50, tol=1e-3)
        page = np.array([pr[v] for v in nodes], dtype=np.float32).reshape(-1, 1)
        del pr
    except Exception:
        page = np.zeros((n, 1), dtype=np.float32)

    # betweenness — sample scales with graph size but stays cheap
    k_bet = min(50 + n // 2000, 150)   # 50 at n=10k … 100 at n=100k
    try:
        bet  = nx.betweenness_centrality(G, k=k_bet, normalized=True, seed=SEED)
        betw = np.array([bet[v] for v in nodes], dtype=np.float32).reshape(-1, 1)
        del bet
    except Exception:
        betw = np.zeros((n, 1), dtype=np.float32)

    gc.collect()
    return StandardScaler().fit_transform(
        np.hstack([deg, clust, tri, page, betw])).astype(np.float32)


def pick_features(G: nx.Graph) -> np.ndarray:
    """Route to the right feature extractor based on graph size."""
    return (get_large_graph_features(G)
            if G.number_of_nodes() > DL_DENSE_LIMIT
            else get_node_features(G))


def embeddings_to_partition(embeddings: np.ndarray,
                             n_communities: int,
                             ground_truth: Optional[Dict[int, int]] = None
                             ) -> Dict[int, int]:
    """
    Cluster node embeddings with K-Means.
    Uses ground-truth k when provided — eliminates the √n heuristic bug.
    Switches to MiniBatchKMeans for n ≥ 5 000.
    """
    n = len(embeddings)
    k = len(set(ground_truth.values())) if ground_truth else n_communities
    k = max(2, min(k, n - 1))
    if n >= 5_000:
        labels = MiniBatchKMeans(
            n_clusters=k, random_state=SEED, n_init=5, batch_size=2048
        ).fit_predict(embeddings)
    else:
        labels = KMeans(
            n_clusters=k, random_state=SEED, n_init=15
        ).fit_predict(embeddings)
    return {i: int(labels[i]) for i in range(n)}
```

## 6. Classical Community Detection Algorithms

**Included (9):** Louvain, Leiden, Walktrap, Infomap, LabelProp, FastLPA, FluidC, MCL, Kernighan–Lin  
**Removed:** Spinglass (exponential complexity on large graphs), LPA-MNI, CLPA (unreliable cdlib dependency)


```python
def louvain_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        import community as community_louvain
        return community_louvain.best_partition(G)
    except Exception:
        return _greedy_fallback(G)


def leiden_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        import leidenalg
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None: return louvain_algorithm(G)
        part = leidenalg.find_partition(ig_graph, leidenalg.ModularityVertexPartition)
        return {nodes[i]: m for i, m in enumerate(part.membership)}
    except Exception:
        return louvain_algorithm(G)


def walktrap_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None: return louvain_algorithm(G)
        comms = ig_graph.community_walktrap().as_clustering()
        return {nodes[i]: m for i, m in enumerate(comms.membership)}
    except Exception:
        return louvain_algorithm(G)


def infomap_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None: return louvain_algorithm(G)
        comms = ig_graph.community_infomap()
        if len(set(comms.membership)) <= 1:
            comms = ig_graph.community_walktrap().as_clustering()
        return {nodes[i]: m for i, m in enumerate(comms.membership)}
    except Exception:
        return walktrap_algorithm(G)


def label_propagation(G: nx.Graph) -> Dict[int, int]:
    """igraph LPA — best of 5 runs."""
    try:
        ig_graph, nodes = _nx_to_igraph(G)
        if ig_graph is None: raise ImportError
        best, best_k = None, 0
        for _ in range(5):
            res = ig_graph.community_label_propagation()
            k   = len(set(res.membership))
            if k > best_k: best_k, best = k, res.membership
        if best and best_k > 1:
            return {nodes[i]: best[i] for i in range(len(nodes))}
        return _greedy_fallback(G)
    except Exception:
        return _greedy_fallback(G)


def label_propagation_fast(G: nx.Graph) -> Dict[int, int]:
    """NetworkX async LPA — best of 5 runs."""
    try:
        from networkx.algorithms.community import asyn_lpa_communities
        best, best_k = None, 0
        for _ in range(5):
            comms = list(asyn_lpa_communities(G))
            if len(comms) > best_k:
                best_k, best = len(comms), _communities_to_partition(comms, G)
        return best if best and best_k > 1 else _greedy_fallback(G)
    except Exception:
        return _greedy_fallback(G)


def fluidc_algorithm(G: nx.Graph) -> Dict[int, int]:
    try:
        from networkx.algorithms.community import asyn_fluidc
        k  = max(2, min(30, int(np.sqrt(G.number_of_nodes()))))
        Gc = (G if nx.is_connected(G)
              else G.subgraph(max(nx.connected_components(G), key=len)).copy())
        return _communities_to_partition(asyn_fluidc(Gc, k), Gc)
    except Exception:
        return _greedy_fallback(G)


def markov_clustering(G: nx.Graph) -> Dict[int, int]:
    if G.number_of_nodes() > 20_000:
        return louvain_algorithm(G)  # MCL intermediates too large above this
    try:
        import markov_clustering as mcl_lib
        import scipy.sparse as sp
        nodes = sorted(G.nodes())
        idx   = {n: i for i, n in enumerate(nodes)}
        n     = len(nodes)
        rows, cols, data = [], [], []
        for u, v in G.edges():
            rows += [idx[u], idx[v]]
            cols += [idx[v], idx[u]]
            data += [1, 1]
        adj      = sp.csr_matrix((data, (rows, cols)), shape=(n, n))
        result   = mcl_lib.run_mcl(adj, inflation=2.0)
        clusters = mcl_lib.get_clusters(result)
        part     = {}
        for cid, cluster in enumerate(clusters):
            for i in cluster: part[nodes[i]] = cid
        return part
    except Exception:
        return _greedy_fallback(G)


def kernighan_lin(G: nx.Graph) -> Dict[int, int]:
    # Recursive bisection only makes sense for small graphs:
    # at n=50k it produces 2^4=16 communities but ground-truth has ~83,
    # and the subgraph copies eat ~300 MB. Fall back to Louvain instead.
    if G.number_of_nodes() > 10_000:
        return louvain_algorithm(G)
    try:
        from networkx.algorithms.community import kernighan_lin_bisection
        def _rkl(sub, depth=0, max_depth=4, offset=0):
            if sub.number_of_nodes() < 10 or depth >= max_depth:
                return {v: offset for v in sub.nodes()}
            try:   p1, p2 = kernighan_lin_bisection(sub)
            except Exception: return {v: offset for v in sub.nodes()}
            out = {}
            out.update(_rkl(sub.subgraph(p1), depth+1, max_depth, offset*2))
            out.update(_rkl(sub.subgraph(p2), depth+1, max_depth, offset*2+1))
            return out
        raw    = _rkl(G)
        unique = {v: i for i, v in enumerate(sorted(set(raw.values())))}
        return {n: unique[l] for n, l in raw.items()}
    except Exception:
        return _greedy_fallback(G)


# ── Registry ──────────────────────────────────────────────────────────────────
CLASSICAL_ALGORITHMS: Dict[str, object] = {
    "Louvain":      louvain_algorithm,
    "Leiden":       leiden_algorithm,
    "Walktrap":     walktrap_algorithm,
    "Infomap":      infomap_algorithm,
    "LabelProp":    label_propagation,
    "FastLPA":      label_propagation_fast,
    "FluidC":       fluidc_algorithm,
    "MCL":          markov_clustering,
    "KernighanLin": kernighan_lin,
}
print(f"Classical algorithms ready: {list(CLASSICAL_ALGORITHMS.keys())}")
```

## 7. Model 1 — GCN (Graph Convolutional Network)

**Reference:** Kipf & Welling, *Semi-Supervised Classification with Graph Convolutional Networks*, ICLR 2017

Architecture: adjacency-smoothed features → Dense(128) → BN → Dropout(0.2) → Dense(64) → BN → embed(16) → reconstruct.  
Parameters: lr=0.005, batch=128, 25 epochs + early stopping (patience=5), `he_uniform` init.


```python
def gcn_community_detection(G: nx.Graph,
                             ground_truth: Optional[Dict[int,int]] = None) -> Dict[int, int]:
    n         = G.number_of_nodes()
    features  = pick_features(G)
    input_dim = features.shape[1]
    EPOCHS    = 25

    if n <= DL_DENSE_LIMIT:
        A              = nx.to_numpy_array(G).astype(np.float32)
        A_hat          = np.eye(n, dtype=np.float32) + A
        D_hat_inv_sqrt = np.diag(1.0 / np.sqrt(np.sum(A_hat, axis=1) + 1e-8))
        A_norm         = (D_hat_inv_sqrt @ A_hat @ D_hat_inv_sqrt).astype(np.float32)
        X_smooth       = (A_norm @ features).astype(np.float32)
    else:
        X_smooth = features   # skip dense A_norm for large graphs

    inp = keras.Input(shape=(input_dim,))
    x   = layers.Dense(128, activation="relu", kernel_initializer="he_uniform")(inp)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(0.2)(x)
    x   = layers.Dense(64, activation="relu")(x)
    x   = layers.BatchNormalization()(x)
    emb = layers.Dense(16, activation="relu", name="embedding")(x)
    out = layers.Dense(input_dim, activation="linear")(emb)

    model = keras.Model(inp, out)
    model.compile(optimizer=Adam(learning_rate=5e-3), loss="mse")

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=5, restore_best_weights=True, verbose=0)
    model.fit(X_smooth, X_smooth, epochs=EPOCHS,
              batch_size=min(2048, max(128, n // 100)),
              verbose=0, shuffle=True, validation_split=0.1,
              callbacks=[TrainingProgress(EPOCHS, "GCN", print_every=5), es])

    node_emb = keras.Model(inp, emb).predict(X_smooth, verbose=0)
    n_comm   = max(2, int(np.sqrt(n)))
    result   = embeddings_to_partition(node_emb, n_comm, ground_truth)
    del model, X_smooth, node_emb
    _clear_keras_session()
    return result
```

## 8. Model 2 — GAT (Graph Attention Network)

**Reference:** Veličković et al., *Graph Attention Networks*, ICLR 2018

Architecture: neighbour-augmented features → Dense(128) → BN → 4 parallel heads(16) →  
Concatenate(64) → BN → Dropout(0.2) → embed(16) → reconstruct. Post-embedding neighbourhood smoothing.  
Parameters: lr=0.005, batch=128, 25 epochs + early stopping (patience=5).


```python
def gat_community_detection(G: nx.Graph,
                             ground_truth: Optional[Dict[int,int]] = None) -> Dict[int, int]:
    n         = G.number_of_nodes()
    features  = pick_features(G)
    input_dim = features.shape[1]
    EPOCHS    = 25

    if n <= DL_DENSE_LIMIT:
        A      = nx.to_numpy_array(G).astype(np.float32)
        deg    = np.sum(A, axis=1, keepdims=True) + 1e-8
        A_norm = A / deg
        X_in   = np.hstack([features, (A_norm @ features)]).astype(np.float32)
    else:
        X_in   = features
        A_norm = None
    aug_dim = X_in.shape[1]

    inp   = keras.Input(shape=(aug_dim,))
    x     = layers.Dense(128, activation="relu", kernel_initializer="he_uniform")(inp)
    x     = layers.BatchNormalization()(x)
    heads = [layers.Dense(16, activation="relu")(x) for _ in range(4)]  # 4×16 = 64
    x     = layers.Concatenate()(heads)
    x     = layers.BatchNormalization()(x)
    x     = layers.Dropout(0.2)(x)
    emb   = layers.Dense(16, activation="relu", name="embedding")(x)
    out   = layers.Dense(aug_dim, activation="linear")(emb)

    model = keras.Model(inp, out)
    model.compile(optimizer=Adam(learning_rate=5e-3), loss="mse")

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=5, restore_best_weights=True, verbose=0)
    model.fit(X_in, X_in, epochs=EPOCHS,
              batch_size=min(2048, max(128, n // 100)),
              verbose=0, shuffle=True, validation_split=0.1,
              callbacks=[TrainingProgress(EPOCHS, "GAT", print_every=5), es])

    node_emb = keras.Model(inp, emb).predict(X_in, verbose=0)
    if A_norm is not None:
        node_emb = 0.5 * node_emb + 0.5 * (A_norm @ node_emb)
    n_comm = max(2, int(np.sqrt(n)))
    result = embeddings_to_partition(node_emb, n_comm, ground_truth)
    del model, X_in, node_emb
    _clear_keras_session()
    return result
```

## 9. Model 3 — VGAE (Variational Graph Autoencoder)

**Reference:** Kipf & Welling, *Variational Graph Auto-Encoders*, NeurIPS Workshop 2016

Architecture: graph-smoothed encoder → μ / log_var heads → reparameterisation (Sampling layer) →  
decoder → MSE reconstruction + β·KL divergence.  
KL warmup ramps β from 0 → 0.5 over the first 10 epochs for stable early training.  
Parameters: embed=16, β_KL=0.5, lr=0.005, batch=128, 30 epochs + early stopping (patience=6).


```python
def vgae_community_detection(G: nx.Graph,
                              ground_truth: Optional[Dict[int,int]] = None) -> Dict[int, int]:
    n         = G.number_of_nodes()
    features  = pick_features(G)
    input_dim = features.shape[1]
    EPOCHS    = 30
    EMBED_DIM = 16

    if n <= DL_DENSE_LIMIT:
        A              = nx.to_numpy_array(G).astype(np.float32)
        A_hat          = np.eye(n, dtype=np.float32) + A
        D_hat_inv_sqrt = np.diag(1.0 / np.sqrt(np.sum(A_hat, axis=1) + 1e-8))
        A_norm         = (D_hat_inv_sqrt @ A_hat @ D_hat_inv_sqrt).astype(np.float32)
        X_smooth       = (A_norm @ features).astype(np.float32)
    else:
        X_smooth = features
        A_norm   = None

    # ── Encoder ───────────────────────────────────────────────────────────────
    enc_in  = keras.Input(shape=(input_dim,), name="enc_in")
    h       = layers.Dense(128, activation="relu",
                            kernel_initializer="he_uniform")(enc_in)
    h       = layers.BatchNormalization()(h)
    h       = layers.Dropout(0.2)(h)
    h       = layers.Dense(64, activation="relu")(h)
    h       = layers.BatchNormalization()(h)
    mu      = layers.Dense(EMBED_DIM, name="mu")(h)
    log_var = layers.Dense(EMBED_DIM, name="log_var")(h)

    class Sampling(layers.Layer):
        def call(self, inputs, training=None):
            mu_, lv_ = inputs
            if training:
                return mu_ + tf.exp(0.5 * lv_) * tf.random.normal(tf.shape(mu_))
            return mu_

    z       = Sampling(name="z")([mu, log_var])
    encoder = keras.Model(enc_in, [mu, log_var, z], name="encoder")

    # ── Decoder ───────────────────────────────────────────────────────────────
    dec_in  = keras.Input(shape=(EMBED_DIM,), name="dec_in")
    decoded = layers.Dense(input_dim, activation="linear", name="decoded")(dec_in)
    decoder = keras.Model(dec_in, decoded, name="decoder")

    # ── KL warmup: ramps β 0 → 0.5 over first 10 epochs ─────────────────────
    beta_var = tf.Variable(0.0, trainable=False, dtype=tf.float32)

    class KLWarmup(keras.callbacks.Callback):
        def on_epoch_begin(self, epoch, logs=None):
            beta_var.assign(min(0.5, epoch / 10.0 * 0.5))

    class VGAEModel(keras.Model):
        def __init__(self, enc, dec):
            super().__init__(); self.enc = enc; self.dec = dec
        def call(self, x, training=None):
            mu_, lv_, z_ = self.enc(x, training=training)
            recon = self.dec(z_, training=training)
            kl    = -0.5 * tf.reduce_mean(1 + lv_ - tf.square(mu_) - tf.exp(lv_))
            self.add_loss(beta_var * kl)
            return recon

    vgae = VGAEModel(encoder, decoder)
    vgae.compile(optimizer=Adam(learning_rate=5e-3), loss="mse")

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=6, restore_best_weights=True, verbose=0)
    vgae.fit(X_smooth, X_smooth, epochs=EPOCHS,
             batch_size=min(2048, max(128, n // 100)),
             verbose=0, shuffle=True, validation_split=0.1,
             callbacks=[TrainingProgress(EPOCHS, "VGAE", print_every=5), es, KLWarmup()])

    mu_out, _, _ = encoder(X_smooth, training=False)
    node_emb     = mu_out.numpy()
    if A_norm is not None:
        node_emb = 0.6 * node_emb + 0.4 * (A_norm @ node_emb)

    n_comm = max(2, int(np.sqrt(n)))
    result = embeddings_to_partition(node_emb, n_comm, ground_truth)
    del vgae, encoder, decoder, X_smooth, node_emb, mu_out
    _clear_keras_session()
    return result
```

## 10. Model 4 — Stacked Autoencoder

**Reference:** Hinton & Salakhutdinov, *Reducing the Dimensionality of Data with Neural Networks*, Science 2006

Architecture: Dense(128) → BN → Dropout(0.2) → Dense(64) → BN → encode(16) → Dense(64) → Dense(128) → reconstruct.  
Parameters: encoding_dim=16, dropout=0.2, lr=0.005, batch=128, 25 epochs + early stopping (patience=5).


```python
def autoencoder_community_detection(G: nx.Graph,
                                     ground_truth: Optional[Dict[int,int]] = None) -> Dict[int, int]:
    n         = G.number_of_nodes()
    features  = pick_features(G)
    input_dim = features.shape[1]
    EPOCHS    = 25

    enc_in  = keras.Input(shape=(input_dim,))
    x       = layers.Dense(128, activation="relu",
                            kernel_initializer="he_uniform")(enc_in)
    x       = layers.BatchNormalization()(x)
    x       = layers.Dropout(0.2)(x)
    x       = layers.Dense(64, activation="relu")(x)
    x       = layers.BatchNormalization()(x)
    encoded = layers.Dense(16, activation="relu", name="encoded")(x)

    x       = layers.Dense(64,        activation="relu")(encoded)
    x       = layers.Dense(128,       activation="relu")(x)
    decoded = layers.Dense(input_dim, activation="linear")(x)

    ae = keras.Model(enc_in, decoded)
    ae.compile(optimizer=Adam(learning_rate=5e-3), loss="mse")

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=5, restore_best_weights=True, verbose=0)
    ae.fit(features, features, epochs=EPOCHS,
           batch_size=min(2048, max(128, n // 100)),
           verbose=0, shuffle=True, validation_split=0.1,
           callbacks=[TrainingProgress(EPOCHS, "Autoencoder", print_every=5), es])

    node_emb = keras.Model(enc_in, encoded).predict(features, verbose=0)
    n_comm   = max(2, int(np.sqrt(n)))
    result   = embeddings_to_partition(node_emb, n_comm, ground_truth)
    del ae, features, node_emb
    _clear_keras_session()
    return result


# ── DL registry ───────────────────────────────────────────────────────────────
DL_ALGORITHMS: Dict[str, object] = {
    "GCN":         gcn_community_detection,
    "GAT":         gat_community_detection,
    "VGAE":        vgae_community_detection,
    "Autoencoder": autoencoder_community_detection,
}
print(f"Deep learning algorithms ready: {list(DL_ALGORITHMS.keys())}")
```

## 11. Benchmark Runner

Wraps every algorithm in `run_with_timeout`.  
DL functions automatically receive the ground-truth partition (for correct k).  
Results include NMI, runtime and status (`ok` / `timeout` / `error`).


```python
def run_benchmark(G: nx.Graph,
                  ground_truth: Dict[int, int],
                  label: str,
                  algorithms: Optional[Dict] = None,
                  timeout_classical: int = TIMEOUT_P1_CLASSICAL,
                  timeout_dl:        int = TIMEOUT_P1_DL) -> pd.DataFrame:
    """
    Run every algorithm on G within its timeout.
    DL algorithms receive ground_truth so they use the correct k.
    Returns a DataFrame with columns:
      Algorithm, Communities, NMI, Runtime, Status
    """
    if algorithms is None:
        algorithms = {**CLASSICAL_ALGORITHMS, **DL_ALGORITHMS}

    k_gt = len(set(ground_truth.values()))
    print(f"\n{'='*72}")
    print(f"  {label}")
    print(f"  Nodes: {G.number_of_nodes():,}  |  Edges: {G.number_of_edges():,}"
          f"  |  GT communities: {k_gt}")
    print(f"{'='*72}")

    rows = []
    for name, fn in algorithms.items():
        is_dl    = name in DL_ALGORITHMS
        t_limit  = timeout_dl if is_dl else timeout_classical
        gt_kwarg = {"ground_truth": ground_truth} if is_dl else {}

        # Skip DL on large graphs where they would OOM
        if is_dl and G.number_of_nodes() > PHASE2_DL_MAX_N:
            print(f"  — {name:<18} | SKIPPED (n>{PHASE2_DL_MAX_N:,} DL memory limit)")
            rows.append({"Algorithm": name, "Communities": np.nan,
                         "NMI": np.nan, "Runtime": np.nan, "Status": "skipped"})
            continue

        try:
            partition, elapsed = run_with_timeout(
                fn, args=(G,), kwargs=gt_kwarg, timeout_sec=t_limit)
            n_comm = len(set(partition.values()))
            nmi    = calculate_nmi(partition, ground_truth)
            icon   = "✓" if nmi > 0.5 else ("~" if nmi > 0.3 else "✗")
            print(f"  {icon} {name:<18} | k={n_comm:4d} | NMI={nmi:.4f}"
                  f" | {elapsed:.1f}s")
            rows.append({"Algorithm": name, "Communities": n_comm,
                         "NMI": round(nmi, 4),
                         "Runtime": round(elapsed, 2), "Status": "ok"})

        except _TimeoutError:
            print(f"  ⏱ {name:<18} | TIMEOUT (>{t_limit}s) — skipped")
            rows.append({"Algorithm": name, "Communities": np.nan,
                         "NMI": np.nan,
                         "Runtime": t_limit, "Status": "timeout"})
        except Exception as e:
            print(f"  ✗ {name:<18} | ERROR: {str(e)[:60]}")
            rows.append({"Algorithm": name, "Communities": np.nan,
                         "NMI": np.nan,
                         "Runtime": np.nan, "Status": "error"})

        # Free any lingering references before the next algorithm
        gc.collect()

    # Release the cached igraph object for this graph
    _clear_igraph_cache()
    gc.collect()
    return pd.DataFrame(rows)
```

## 12. Phase 1 — Small & Medium Graphs (n = 500, 1 000, 5 000)

All 13 algorithms run on every (n, μ) combination.  
Results are saved to CSV after each configuration.


```python
PHASE1_SIZES = [500, 1_000, 5_000]
MU_VALUES    = [0.3, 0.5, 0.7]

phase1_results: List[pd.DataFrame] = []

print("\n" + "="*72)
print("PHASE 1 — Full Benchmark  (n ∈ {500, 1 000, 5 000})")
print("="*72)

for n in PHASE1_SIZES:
    for mu in MU_VALUES:
        label = f"LFR  n={n:,}  μ={mu}"
        try:
            G, gt = generate_lfr_graph(n=n, mu=mu)
            df    = run_benchmark(G, gt, label,
                                  timeout_classical=TIMEOUT_P1_CLASSICAL,
                                  timeout_dl=TIMEOUT_P1_DL)
            df["n"]     = n
            df["mu"]    = mu
            df["phase"] = 1
            phase1_results.append(df)
        except Exception as e:
            print(f"  ✗ FAILED n={n}, μ={mu}: {e}")

phase1_df = (pd.concat(phase1_results, ignore_index=True)
             if phase1_results else pd.DataFrame())
if not phase1_df.empty:
    print("\n✓ Phase 1 complete")
```

## 13. Algorithm Selection — Promoting Top Performers to Phase 2

Algorithms are ranked by **average NMI** across all Phase 1 configurations.  
The top `TOP_N_PHASE2` algorithms proceed to Phase 2. Louvain is always kept as baseline.


```python
# ── Phase 2 algorithm showcase — fixed set (independent of Phase 1 results) ──
# Louvain, Leiden, Infomap, FluidC are the algorithms showcased in Phase 2.
PHASE2_FIXED_ALGORITHMS = ["Louvain", "Leiden", "Infomap", "FluidC"]

all_fns = {**CLASSICAL_ALGORITHMS, **DL_ALGORITHMS}
phase2_algorithms = {name: all_fns[name]
                     for name in PHASE2_FIXED_ALGORITHMS
                     if name in all_fns}

# Still show Phase 1 ranking for reference (informational only)
if not phase1_df.empty:
    ranking = (phase1_df[phase1_df["Status"] == "ok"]
               .groupby("Algorithm")["NMI"]
               .mean().dropna()
               .sort_values(ascending=False))
    print("\nPhase 1 full ranking (avg NMI across all n & μ) — for reference:")
    print("-" * 50)
    for i, (alg, score) in enumerate(ranking.items(), 1):
        kind = " [DL]" if alg in DL_ALGORITHMS else "     "
        tag  = "  ◀ PHASE 2" if alg in PHASE2_FIXED_ALGORITHMS else ""
        print(f"  {i:2d}.{kind} {alg:<18} NMI={score:.4f}{tag}")

print(f"\n→ Phase 2 algorithms (fixed showcase): {list(phase2_algorithms.keys())}")
```

## 14. Phase 2 — Large Graphs (n = 10 000, 50 000, 100 000)

Fixed showcase algorithms (**Louvain, Leiden, Infomap, FluidC**) run on all Phase 2 sizes.  
For n = 50 000 and n = 100 000 only μ = 0.3 and μ = 0.5 are tested (μ = 0.7 skipped — too noisy at scale).  
Timeouts are extended to handle the larger graphs.

```python
PHASE2_SIZES = [10_000, 50_000, 100_000]

# μ = 0.7 is skipped for n ≥ 50 000 (LFR convergence is unreliable at scale)
PHASE2_MU_LARGE = [0.3, 0.5]   # used for n = 50 000 and n = 100 000
PHASE2_MU_ALL   = [0.3, 0.5, 0.7]  # used for n = 10 000

phase2_results: List[pd.DataFrame] = []

print("\n" + "="*72)
print("PHASE 2 — Large Graphs  (n ∈ {10 000, 50 000, 100 000})")
print(f"Algorithms: {list(phase2_algorithms.keys())}")
print("μ values: n=10k → [0.3, 0.5, 0.7]  |  n=50k, 100k → [0.3, 0.5]")
print("="*72)

for n in PHASE2_SIZES:
    mu_list = PHASE2_MU_ALL if n == 10_000 else PHASE2_MU_LARGE
    for mu in mu_list:
        label = f"LFR  n={n:,}  μ={mu}"
        G, gt = None, None
        try:
            # ── Timeout guard for graph generation ───────────────────────
            G, gt = run_with_timeout(
                generate_lfr_graph,
                kwargs={"n": n, "mu": mu},
                timeout_sec=TIMEOUT_P2_GRAPH,
            )[0]  # returns (result, elapsed); unpack result tuple
            df    = run_benchmark(G, gt, label,
                                  algorithms=phase2_algorithms,
                                  timeout_classical=TIMEOUT_P2_CLASSICAL,
                                  timeout_dl=TIMEOUT_P2_DL)
            df["n"]     = n
            df["mu"]    = mu
            df["phase"] = 2
            phase2_results.append(df)
        except _TimeoutError:
            print(f"  ⏱ GRAPH-GEN TIMEOUT n={n:,}, μ={mu} "
                  f"(>{TIMEOUT_P2_GRAPH}s) — skipped")
        except Exception as e:
            print(f"  ✗ FAILED n={n}, μ={mu}: {e}")
        finally:
            # Always release the graph object and force GC before next config
            if G is not None: del G
            if gt is not None: del gt
            gc.collect()

phase2_df = (pd.concat(phase2_results, ignore_index=True)
             if phase2_results else pd.DataFrame())
if not phase2_df.empty:
    print("\n✓ Phase 2 complete")

# ── Save single combined CSV (both phases) ────────────────────────────────────
all_results = []
if not phase1_df.empty: all_results.append(phase1_df)
if not phase2_df.empty: all_results.append(phase2_df)

if all_results:
    combined_df = pd.concat(all_results, ignore_index=True)
    combined_df.to_csv("lfr_benchmark_results.csv", index=False)
    print(f"\n✓ All results saved → lfr_benchmark_results.csv")
    print(f"  Total rows: {len(combined_df):,}")
    print(f"  Columns   : {list(combined_df.columns)}")
```

## 15. Visualizations

Four plots are produced:
1. **Phase 1 grid** — NMI bar charts per (n, μ) in a 3×3 layout
2. **Phase 1 heatmap** — algorithm × configuration NMI matrix
3. **Phase 2 scalability** — NMI vs n line plot per μ
4. **Overall ranking** — composite bar chart across both phases


```python
ALGO_COLORS = {
    # Classical
    "Louvain":      "#4C72B0", "Leiden":       "#55A868",
    "Walktrap":     "#C44E52", "Infomap":      "#8172B2",
    "LabelProp":    "#CCB974", "FastLPA":      "#64B5CD",
    "FluidC":       "#E08900", "MCL":          "#A9A9A9",
    "KernighanLin": "#D16B6B",
    # Deep learning
    "GCN":          "#1F77B4", "GAT":          "#FF7F0E",
    "VGAE":         "#2CA02C", "Autoencoder":  "#D62728",
}


def _bar_panel(ax, avg_df, title):
    """Horizontal bar chart for a single (n, μ) panel."""
    best   = avg_df.idxmax()
    colors = [ALGO_COLORS.get(a, "#888888") for a in avg_df.index]
    bars   = ax.barh(avg_df.index, avg_df.values, color=colors,
                     edgecolor="white", linewidth=0.8)
    for bar, alg in zip(bars, avg_df.index):
        if alg == best:
            bar.set_edgecolor("black"); bar.set_linewidth(2)
    for bar, val in zip(bars, avg_df.values):
        ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
                f"{val:.3f}", va="center", fontsize=8)
    ax.set_xlim(0, 1.1)
    ax.set_xlabel("NMI", fontsize=9)
    ax.set_title(title, fontsize=10, fontweight="bold")
    ax.text(0.97, 0.03, f"★ {best}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=8, fontweight="bold",
            bbox=dict(boxstyle="round,pad=0.3", fc="gold", alpha=0.85))


# ── 1. Phase 1 grid ───────────────────────────────────────────────────────────
def plot_phase1_grid(df: pd.DataFrame):
    sizes = sorted(df["n"].unique())
    mus   = sorted(df["mu"].unique())
    fig, axes = plt.subplots(len(sizes), len(mus),
                             figsize=(6*len(mus), 5*len(sizes)))
    if len(sizes) == 1: axes = axes[np.newaxis, :]
    if len(mus)   == 1: axes = axes[:, np.newaxis]

    for i, n in enumerate(sizes):
        for j, mu in enumerate(mus):
            ax  = axes[i, j]
            sub = df[(df["n"]==n) & (df["mu"]==mu) & (df["Status"]=="ok")]
            if sub.empty:
                ax.text(0.5, 0.5, "No data", ha="center", va="center")
                ax.set_title(f"n={n:,}, μ={mu}"); continue
            avg = sub.groupby("Algorithm")["NMI"].mean().sort_values()
            _bar_panel(ax, avg, f"n={n:,}  μ={mu}")

    plt.suptitle("Phase 1 — NMI by Algorithm (rows: graph size  |  cols: mixing parameter μ)",fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("lfr_phase1_grid.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved: lfr_phase1_grid.png")


# ── 2. Phase 1 heatmap ────────────────────────────────────────────────────────
def plot_phase1_heatmap(df: pd.DataFrame):
    ok = df[df["Status"] == "ok"].copy()
    ok["config"] = ok.apply(lambda r: f"n={int(r['n']):,}μ={r['mu']}", axis=1)
    pivot = ok.pivot_table(index="Algorithm", columns="config",values="NMI", aggfunc="mean")
    fig, ax = plt.subplots(figsize=(max(10, len(pivot.columns)*1.5),max(6, len(pivot)*0.6)))
    sns.heatmap(pivot, annot=True, fmt=".3f", cmap="RdYlGn",
                vmin=0, vmax=1, linewidths=0.4, linecolor="gray",
                cbar_kws={"label": "NMI"}, ax=ax)
    ax.set_title("Phase 1 — NMI Heatmap (Algorithm × Configuration)",
                 fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig("lfr_phase1_heatmap.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved: lfr_phase1_heatmap.png")


# ── 3. Phase 2 scalability ────────────────────────────────────────────────────
def plot_phase2_scalability(df: pd.DataFrame):
    mus  = sorted(df["mu"].unique())
    fig, axes = plt.subplots(1, len(mus), figsize=(7*len(mus), 6), sharey=True)
    if len(mus) == 1: axes = [axes]

    for ax, mu in zip(axes, mus):
        sub = df[(df["mu"] == mu) & (df["Status"] == "ok")]
        for alg, grp in sub.groupby("Algorithm"):
            grp_s = grp.sort_values("n")
            ax.plot(grp_s["n"], grp_s["NMI"], marker="o", label=alg,
                    color=ALGO_COLORS.get(alg, "#888888"))
        ax.set_xscale("log")
        ax.set_xlabel("Graph size (n)", fontsize=11)
        ax.set_ylabel("NMI", fontsize=11)
        ax.set_ylim(0, 1)
        ax.set_title(f"μ = {mu}", fontsize=12, fontweight="bold")
        ax.legend(fontsize=8)
        ax.xaxis.set_major_formatter(
            ticker.FuncFormatter(lambda v, _: f"{int(v):,}"))

    plt.suptitle("Phase 2 — NMI Scalability  (n = 10k, 50k, 100k)",
                 fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("lfr_phase2_scalability.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved: lfr_phase2_scalability.png")


# ── 4. Overall ranking ────────────────────────────────────────────────────────
def plot_overall_ranking(p1: pd.DataFrame, p2: pd.DataFrame):
    combined = (pd.concat([p1, p2], ignore_index=True)
                if not p2.empty else p1.copy())
    ranking  = (combined[combined["Status"] == "ok"]
                .groupby("Algorithm")["NMI"]
                .mean().dropna()
                .sort_values(ascending=False))

    fig, ax = plt.subplots(figsize=(10, max(6, len(ranking) * 0.5)))
    colors  = [ALGO_COLORS.get(a, "#888888") for a in ranking.index]
    bars    = ax.barh(ranking.index[::-1], ranking.values[::-1],
                      color=colors[::-1], edgecolor="white", linewidth=0.8)
    # Gold border on winner
    bars[0].set_edgecolor("black"); bars[0].set_linewidth(2.5)
    for bar, val in zip(bars, ranking.values[::-1]):
        ax.text(bar.get_width() + 0.005, bar.get_y() + bar.get_height()/2,
                f"{val:.4f}", va="center", fontsize=9)
    ax.set_xlim(0, 1.1)
    ax.set_xlabel("Average NMI (all configurations)", fontsize=12)
    ax.set_title("Overall Algorithm Ranking — LFR Benchmark",
                 fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("lfr_overall_ranking.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved: lfr_overall_ranking.png")


# ── Run all plots ─────────────────────────────────────────────────────────────
if not phase1_df.empty:
    plot_phase1_grid(phase1_df)
    plot_phase1_heatmap(phase1_df)

if not phase2_df.empty:
    plot_phase2_scalability(phase2_df)

p2_for_rank = phase2_df if not phase2_df.empty else pd.DataFrame()
if not phase1_df.empty:
    plot_overall_ranking(phase1_df, p2_for_rank)
```

## 16. Final Summary

```python
def print_summary(p1: pd.DataFrame, p2: pd.DataFrame):
    combined = (pd.concat([p1, p2], ignore_index=True)
                if not p2.empty else p1.copy())
    ok = combined[combined["Status"] == "ok"]

    print("\n" + "="*72)
    print("FINAL RANKING — Average NMI (Phase 1 + Phase 2)")
    print("="*72)
    ranking = (ok.groupby("Algorithm")["NMI"]
               .mean().dropna().sort_values(ascending=False))
    for i, (alg, score) in enumerate(ranking.items(), 1):
        kind = "[DL] " if alg in DL_ALGORITHMS else "[CLS]"
        tag  = "  ◀ WINNER" if i == 1 else ("  TOP 3" if i <= 3 else "")
        print(f"  {i:2d}. {kind} {alg:<18} NMI={score:.4f}{tag}")

    print("\n" + "="*72)
    print("PHASE 1 — Best algorithm per μ")
    print("="*72)
    for mu in sorted(ok["mu"].dropna().unique()):
        sub   = ok[ok["mu"] == mu]
        best  = sub.groupby("Algorithm")["NMI"].mean().idxmax()
        score = sub.groupby("Algorithm")["NMI"].mean().max()
        print(f"  μ={mu}  →  {best:<18} NMI={score:.4f}")

    if not p2.empty:
        ok2 = p2[p2["Status"] == "ok"]
        print("\n" + "="*72)
        print("PHASE 2 — Best algorithm per graph size")
        print("="*72)
        for n in sorted(ok2["n"].dropna().unique()):
            sub   = ok2[ok2["n"] == n]
            best  = sub.groupby("Algorithm")["NMI"].mean().idxmax()
            score = sub.groupby("Algorithm")["NMI"].mean().max()
            print(f"  n={int(n):>7,}  →  {best:<18} NMI={score:.4f}")

    print("\n" + "="*72)
    print("TIMEOUT / ERROR LOG")
    print("="*72)
    bad = combined[combined["Status"] != "ok"]
    if bad.empty:
        print("  All algorithms completed within timeout on every graph.")
    else:
        for _, row in bad[["Algorithm","n","mu","Status"]].iterrows():
            print(f"  {row.Algorithm:<18} n={int(row['n']):,}  "
                  f"μ={row['mu']}  → {row['Status']}")
    print("="*72)


if not phase1_df.empty:
    print_summary(phase1_df, p2_for_rank)
```
