# Hyperparameter Tuning — ALS & BM25
Grid search with leave-some-out evaluation (Precision@5, Recall@5, F1@5).
All helper classes (ALSRecommender, BM25Recommender, etc.) are assumed to be
available from the earlier notebook cells.

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T1 — Shared leave-some-out evaluator (reusable for any model)
# ══════════════════════════════════════════════════════════════════════════
import numpy as np
import pandas as pd
import itertools, time, warnings
warnings.filterwarnings('ignore')

def evaluate_model(recommender, user_item_matrix,
                   k_values=(5, 10), n_users=300, random_state=42):
    """
    Leave-some-out evaluation: split each user's interactions 50/50 into
    train / test, hide test items, measure Precision / Recall / F1 @ k.

    Parameters
    ----------
    recommender : object with .recommend(user_idx, masked_matrix, top_n) -> list[(item_idx, score)]
    user_item_matrix : scipy.sparse CSR matrix
    k_values : iterable of int
    n_users : int — how many users to sample (first n, for speed)
    random_state : int

    Returns
    -------
    dict with keys 'precision@k', 'recall@k', 'f1@k'
    """
    rng = np.random.RandomState(random_state)
    results = {k: {'precision': [], 'recall': []} for k in k_values}

    for user_idx in range(min(n_users, user_item_matrix.shape[0])):
        user_items = list(user_item_matrix[user_idx].nonzero()[1])
        if len(user_items) < 3:
            continue

        n_train = max(1, len(user_items) // 2)
        perm = rng.permutation(len(user_items))
        train_items = set(np.array(user_items)[perm[:n_train]])
        test_items  = set(np.array(user_items)[perm[n_train:]])
        if not test_items:
            continue

        masked = user_item_matrix.tolil()
        for i in test_items:
            masked[user_idx, i] = 0
        masked = masked.tocsr()

        for k in k_values:
            recs = recommender.recommend(user_idx, masked, top_n=k)
            recommended = set(idx for idx, _ in recs)
            hits = len(recommended & test_items)
            results[k]['precision'].append(hits / len(recommended) if recommended else 0)
            results[k]['recall'].append(hits / len(test_items))

    out = {}
    for k in k_values:
        p = np.mean(results[k]['precision']) if results[k]['precision'] else 0
        r = np.mean(results[k]['recall'])    if results[k]['recall']    else 0
        f = 2*p*r/(p+r) if (p+r) > 0 else 0
        out[f'precision@{k}'] = round(p, 5)
        out[f'recall@{k}']    = round(r, 5)
        out[f'f1@{k}']        = round(f, 5)
    return out

print('evaluate_model() ready.')
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T2 — Thin wrappers so every model exposes .recommend()
# ══════════════════════════════════════════════════════════════════════════

class ALSWrapper:
    """
    Wraps ALSRecommender so it exposes the .recommend() interface
    expected by evaluate_model().
    """
    def __init__(self, n_factors=20, n_iterations=10,
                 regularization=0.1, alpha=40, random_state=42):
        self.params = dict(
            n_factors=n_factors, n_iterations=n_iterations,
            regularization=regularization, alpha=alpha,
            random_state=random_state
        )
        self._model = None

    def fit(self, user_item_matrix):
        self._model = ALSRecommender(**self.params)
        self._model.fit(user_item_matrix)
        return self

    def recommend(self, user_idx, user_item_matrix, top_n=10):
        all_scores   = self._model.user_factors[user_idx].dot(self._model.item_factors.T)
        already_bought = set(user_item_matrix[user_idx].nonzero()[1])
        candidates   = [i for i in range(len(all_scores)) if i not in already_bought]
        if not candidates:
            return []
        scores = all_scores[candidates]
        order  = np.argsort(scores)[::-1][:top_n]
        return [(candidates[i], float(scores[i])) for i in order]


class BM25Wrapper:
    """
    Wraps BM25Recommender so it exposes the .recommend() interface.
    """
    def __init__(self, k1=1.5, b=0.75):
        self.params = dict(k1=k1, b=b)
        self._model = None

    def fit(self, user_item_matrix):
        self._model = BM25Recommender(**self.params)
        self._model.fit(user_item_matrix)
        return self

    def recommend(self, user_idx, user_item_matrix, top_n=10):
        seed_items = list(user_item_matrix[user_idx].nonzero()[1])
        if not seed_items or self._model.item_scores is None:
            return []
        scores = np.zeros(self._model.item_scores.shape[1], dtype=np.float32)
        for item_idx in seed_items:
            rating = user_item_matrix[user_idx, item_idx]
            scores += rating * self._model.item_scores[item_idx]
        already_bought = set(seed_items)
        candidates = [i for i in range(len(scores)) if i not in already_bought]
        if not candidates:
            return []
        cand_scores = scores[candidates]
        order = np.argsort(cand_scores)[::-1][:top_n]
        return [(candidates[i], float(cand_scores[i])) for i in order]


print('ALSWrapper and BM25Wrapper ready.')
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T3 — ALS Grid Search
#
#  Tuned parameters:
#    n_factors      — latent space dimensionality (expressiveness vs overfitting)
#    regularization — L2 penalty on user/item factors (prevents overfitting)
#    alpha          — confidence scaling for implicit feedback (0-rating weight)
#    n_iterations   — ALS solver passes (convergence)
#
#  Primary metric: F1@5  (balances precision and recall at top-5)
# ══════════════════════════════════════════════════════════════════════════

ALS_GRID = {
    'n_factors':      [10, 20, 50, 100],
    'regularization': [0.01, 0.1, 0.5, 1.0],
    'alpha':          [10, 40, 80],
    'n_iterations':   [10, 20],
}

als_combos = list(itertools.product(
    ALS_GRID['n_factors'],
    ALS_GRID['regularization'],
    ALS_GRID['alpha'],
    ALS_GRID['n_iterations']
))
print(f'ALS grid: {len(als_combos)} combinations')

als_tuning_rows = []

for idx, (n_f, reg, alph, n_it) in enumerate(als_combos, 1):
    t0 = time.time()
    try:
        model = ALSWrapper(
            n_factors=n_f, n_iterations=n_it,
            regularization=reg, alpha=alph
        ).fit(user_item_matrix)
        metrics = evaluate_model(model, user_item_matrix, n_users=300)
    except Exception as e:
        metrics = {'precision@5': 0, 'recall@5': 0, 'f1@5': 0,
                   'precision@10': 0, 'recall@10': 0, 'f1@10': 0}
        print(f'  [WARN] combo {idx} failed: {e}')

    elapsed = round(time.time() - t0, 1)
    row = dict(n_factors=n_f, regularization=reg, alpha=alph, n_iterations=n_it,
               elapsed_s=elapsed, **metrics)
    als_tuning_rows.append(row)

    if idx % 10 == 0 or idx == len(als_combos):
        current_best = max(als_tuning_rows, key=lambda r: r['f1@5'])
        print(f'[{idx:>3}/{len(als_combos)}] ')
        print(f'  last: n_factors={n_f}, reg={reg}, alpha={alph}, n_it={n_it}'
              f' → F1@5={metrics["f1@5"]:.5f}  ({elapsed}s)')
        print(f'  best so far: n_factors={current_best["n_factors"]}, '
              f'reg={current_best["regularization"]}, '
              f'alpha={current_best["alpha"]}, '
              f'n_it={current_best["n_iterations"]} '
              f'→ F1@5={current_best["f1@5"]:.5f}')

als_df = pd.DataFrame(als_tuning_rows).sort_values('f1@5', ascending=False).reset_index(drop=True)
print('\n=== ALS Tuning Complete ===')
print(als_df.head(10).to_string(index=False))
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T4 — BM25 Grid Search
#
#  Tuned parameters:
#    k1 — term frequency saturation: higher = slower TF saturation
#    b  — length normalisation: 0 = no normalisation, 1 = full normalisation
#
#  BM25 fits and evaluates very quickly → fine grid is feasible
# ══════════════════════════════════════════════════════════════════════════

BM25_GRID = {
    'k1': [0.5, 1.0, 1.2, 1.5, 2.0, 3.0],
    'b':  [0.0, 0.25, 0.5, 0.75, 1.0],
}

bm25_combos = list(itertools.product(BM25_GRID['k1'], BM25_GRID['b']))
print(f'BM25 grid: {len(bm25_combos)} combinations')

bm25_tuning_rows = []

for idx, (k1, b) in enumerate(bm25_combos, 1):
    t0 = time.time()
    try:
        model = BM25Wrapper(k1=k1, b=b).fit(user_item_matrix)
        metrics = evaluate_model(model, user_item_matrix, n_users=300)
    except Exception as e:
        metrics = {'precision@5': 0, 'recall@5': 0, 'f1@5': 0,
                   'precision@10': 0, 'recall@10': 0, 'f1@10': 0}
        print(f'  [WARN] k1={k1}, b={b} failed: {e}')

    elapsed = round(time.time() - t0, 1)
    row = dict(k1=k1, b=b, elapsed_s=elapsed, **metrics)
    bm25_tuning_rows.append(row)
    print(f'[{idx:>2}/{len(bm25_combos)}] k1={k1}, b={b} → F1@5={metrics["f1@5"]:.5f}  ({elapsed}s)')

bm25_df = pd.DataFrame(bm25_tuning_rows).sort_values('f1@5', ascending=False).reset_index(drop=True)
print('\n=== BM25 Tuning Complete ===')
print(bm25_df.to_string(index=False))
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T5 — Sensitivity analysis heatmaps (BM25)
# ══════════════════════════════════════════════════════════════════════════
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('BM25 Hyperparameter Sensitivity (F1@5, Precision@5, Recall@5)', fontsize=13)

for ax, metric in zip(axes, ['f1@5', 'precision@5', 'recall@5']):
    pivot = bm25_df.pivot_table(index='k1', columns='b', values=metric, aggfunc='max')
    sns.heatmap(pivot, annot=True, fmt='.4f', cmap='YlGnBu',
                linewidths=0.5, linecolor='grey', ax=ax,
                cbar_kws={'shrink': 0.8})
    ax.set_title(metric)
    ax.set_xlabel('b (length norm)')
    ax.set_ylabel('k1 (TF saturation)')

plt.tight_layout()
plt.savefig('bm25_sensitivity.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved bm25_sensitivity.png')
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T6 — ALS sensitivity: n_factors vs regularization
# ══════════════════════════════════════════════════════════════════════════

# Pivot over (n_factors, regularization), averaging across alpha / n_iterations
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('ALS Sensitivity: n_factors × regularization (avg over alpha, n_iter)', fontsize=13)

for ax, metric in zip(axes, ['f1@5', 'precision@5', 'recall@5']):
    pivot = als_df.pivot_table(
        index='n_factors', columns='regularization', values=metric, aggfunc='max'
    )
    sns.heatmap(pivot, annot=True, fmt='.4f', cmap='YlOrRd',
                linewidths=0.5, linecolor='grey', ax=ax,
                cbar_kws={'shrink': 0.8})
    ax.set_title(metric)
    ax.set_xlabel('regularization (λ)')
    ax.set_ylabel('n_factors')

plt.tight_layout()
plt.savefig('als_sensitivity_factors_reg.png', dpi=150, bbox_inches='tight')
plt.show()

# Pivot over (alpha, n_iterations), best-case per pair
fig2, axes2 = plt.subplots(1, 3, figsize=(18, 5))
fig2.suptitle('ALS Sensitivity: alpha × n_iterations (best over n_factors, reg)', fontsize=13)

for ax, metric in zip(axes2, ['f1@5', 'precision@5', 'recall@5']):
    pivot = als_df.pivot_table(
        index='alpha', columns='n_iterations', values=metric, aggfunc='max'
    )
    sns.heatmap(pivot, annot=True, fmt='.4f', cmap='BuGn',
                linewidths=0.5, linecolor='grey', ax=ax,
                cbar_kws={'shrink': 0.8})
    ax.set_title(metric)
    ax.set_xlabel('n_iterations')
    ax.set_ylabel('alpha')

plt.tight_layout()
plt.savefig('als_sensitivity_alpha_nit.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved als_sensitivity_factors_reg.png and als_sensitivity_alpha_nit.png')
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T7 — Extract best configs & re-fit final tuned models
# ══════════════════════════════════════════════════════════════════════════

# --- Best ALS ---
best_als_row = als_df.iloc[0]
print('=== Best ALS Configuration ===')
print(best_als_row[['n_factors','regularization','alpha','n_iterations',
                     'f1@5','precision@5','recall@5']].to_string())

best_als = ALSWrapper(
    n_factors      = int(best_als_row['n_factors']),
    n_iterations   = int(best_als_row['n_iterations']),
    regularization = float(best_als_row['regularization']),
    alpha          = float(best_als_row['alpha']),
).fit(user_item_matrix)

# Expose inner model so downstream code (ALSCommunityRecommender etc.) can use it
als_rec_tuned = best_als._model
print(f'\nTuned ALS re-fitted. user_factors: {als_rec_tuned.user_factors.shape}')

# --- Best BM25 ---
best_bm25_row = bm25_df.iloc[0]
print('\n=== Best BM25 Configuration ===')
print(best_bm25_row[['k1','b','f1@5','precision@5','recall@5']].to_string())

best_bm25 = BM25Wrapper(
    k1 = float(best_bm25_row['k1']),
    b  = float(best_bm25_row['b']),
).fit(user_item_matrix)

bm25_rec_tuned = best_bm25._model
print(f'Tuned BM25 re-fitted. item_scores: {bm25_rec_tuned.item_scores.shape}')
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T8 — Compare default vs tuned models
# ══════════════════════════════════════════════════════════════════════════

# Default models from original notebook
default_als  = ALSWrapper(n_factors=20, n_iterations=10,
                           regularization=0.1, alpha=40).fit(user_item_matrix)
default_bm25 = BM25Wrapper(k1=1.5, b=0.75).fit(user_item_matrix)

default_als_metrics  = evaluate_model(default_als,  user_item_matrix, n_users=300)
default_bm25_metrics = evaluate_model(default_bm25, user_item_matrix, n_users=300)
tuned_als_metrics    = evaluate_model(best_als,     user_item_matrix, n_users=300)
tuned_bm25_metrics   = evaluate_model(best_bm25,    user_item_matrix, n_users=300)

comparison = pd.DataFrame([
    {'Model': 'ALS (default)',  **default_als_metrics},
    {'Model': 'ALS (tuned)',    **tuned_als_metrics},
    {'Model': 'BM25 (default)', **default_bm25_metrics},
    {'Model': 'BM25 (tuned)',   **tuned_bm25_metrics},
]).set_index('Model')

print('=== Default vs Tuned ===')
print(comparison.to_string())

# Compute deltas
als_delta_f1  = tuned_als_metrics['f1@5']  - default_als_metrics['f1@5']
bm25_delta_f1 = tuned_bm25_metrics['f1@5'] - default_bm25_metrics['f1@5']
print(f'\nALS  F1@5 improvement : {als_delta_f1:+.5f} '
      f'({als_delta_f1/max(default_als_metrics["f1@5"], 1e-9)*100:+.1f}%)')
print(f'BM25 F1@5 improvement : {bm25_delta_f1:+.5f} '
      f'({bm25_delta_f1/max(default_bm25_metrics["f1@5"], 1e-9)*100:+.1f}%)')

# Bar chart
import matplotlib.pyplot as plt
import numpy as np

labels  = comparison.index.tolist()
metrics = ['precision@5', 'recall@5', 'f1@5']
colors  = ['#4C72B0','#55A868','#DD8452','#C44E52']
x = np.arange(len(labels))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Default vs Tuned — ALS & BM25', fontsize=13, fontweight='bold')

for ax, metric in zip(axes, metrics):
    vals = comparison[metric].values
    bars = ax.bar(x, vals, color=colors, width=0.6, edgecolor='white')
    ax.set_xticks(x)
    ax.set_xticklabels(labels, rotation=15, ha='right', fontsize=9)
    ax.set_title(metric, fontsize=11)
    ax.grid(axis='y', alpha=0.3)
    for bar, v in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.0003,
                f'{v:.4f}', ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.savefig('als_bm25_tuning_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved als_bm25_tuning_comparison.png')
```

```python
# ══════════════════════════════════════════════════════════════════════════
#  SNIPPET T9 — Save tuned results & update model bundle
# ══════════════════════════════════════════════════════════════════════════
import os, pickle

# Save tuning tables
als_df.to_csv('als_tuning_results.csv', index=False)
bm25_df.to_csv('bm25_tuning_results.csv', index=False)
print('Saved als_tuning_results.csv and bm25_tuning_results.csv')

# Patch existing model bundle if it exists
MODEL_DIR   = 'recommender_models'
bundle_path = os.path.join(MODEL_DIR, 'models.pkl')

if os.path.exists(bundle_path):
    with open(bundle_path, 'rb') as f:
        bundle = pickle.load(f)

    bundle['als']        = als_rec_tuned    # replace with tuned version
    bundle['bm25']       = bm25_rec_tuned   # replace with tuned version
    bundle['als_best_params']  = best_als_row[['n_factors','regularization',
                                                'alpha','n_iterations']].to_dict()
    bundle['bm25_best_params'] = best_bm25_row[['k1','b']].to_dict()
    bundle['tuning_metrics']   = {
        'ALS_default':  default_als_metrics,
        'ALS_tuned':    tuned_als_metrics,
        'BM25_default': default_bm25_metrics,
        'BM25_tuned':   tuned_bm25_metrics,
    }

    with open(bundle_path, 'wb') as f:
        pickle.dump(bundle, f, protocol=4)
    print(f'Updated {bundle_path} with tuned ALS + BM25 models.')
else:
    print(f'Bundle not found at {bundle_path} — run the original notebook first, '
          'then re-run this cell.')

print('\n=== Final Best Hyperparameters ===')
print('ALS  :', {k: best_als_row[k] for k in ['n_factors','regularization','alpha','n_iterations']})
print('BM25 :', {'k1': best_bm25_row['k1'], 'b': best_bm25_row['b']})
```
