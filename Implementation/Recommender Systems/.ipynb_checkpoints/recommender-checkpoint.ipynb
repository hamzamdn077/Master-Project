```python
# Snippet 1: Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.sparse import csr_matrix
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split
import networkx as nx
from scipy.spatial.distance import pdist, squareform
import warnings
warnings.filterwarnings('ignore')
import community as community_louvain
import leidenalg as la
import igraph as ig
from cdlib import algorithms
from collections import defaultdict

print("All libraries imported successfully!")
```

```python
df = pd.read_excel("Online Retail.xlsx")
```

```python
print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
df.head()
```

```python
# Snippet 3: Clean and prepare data
def preprocess_data(df):
    # Remove cancelled orders
    df = df[~df['InvoiceNo'].astype(str).str.contains('C', na=False)]
    
    # Remove missing values
    df = df.dropna(subset=['CustomerID', 'StockCode'])
    
    # Filter out negative quantities
    df = df[df['Quantity'] > 0]
    
    # Filter out zero price items
    df = df[df['UnitPrice'] > 0]
    
    # Create user-item rating matrix (using purchase count as rating)
    df_purchases = df.groupby(['CustomerID', 'StockCode']).size().reset_index(name='Rating')
    
    # Cap ratings at 5 for simplicity
    df_purchases['Rating'] = df_purchases['Rating'].clip(upper=5)
    
    return df_purchases

df_purchases = preprocess_data(df)
print(f"Processed data shape: {df_purchases.shape}")
print(f"Number of unique users: {df_purchases['CustomerID'].nunique()}")
print(f"Number of unique items: {df_purchases['StockCode'].nunique()}")
print(f"\nRating distribution:")
print(df_purchases['Rating'].value_counts().sort_index())
```

```python
# Snippet 4: Create user-item matrix with sparse representation
from scipy.sparse import lil_matrix

def create_user_item_matrix(df_purchases):
    # Get unique users and items
    users = df_purchases['CustomerID'].unique()
    items = df_purchases['StockCode'].unique()
    
    # Create mappings
    user_to_idx = {user: idx for idx, user in enumerate(users)}
    item_to_idx = {item: idx for idx, item in enumerate(items)}
    
    # Create sparse matrix
    n_users = len(users)
    n_items = len(items)
    matrix = lil_matrix((n_users, n_items), dtype=np.float32)
    
    # Fill matrix
    for _, row in df_purchases.iterrows():
        user_idx = user_to_idx[row['CustomerID']]
        item_idx = item_to_idx[row['StockCode']]
        matrix[user_idx, item_idx] = row['Rating']
    
    return matrix, users, items, user_to_idx, item_to_idx

user_item_matrix, users, items, user_to_idx, item_to_idx = create_user_item_matrix(df_purchases)
print(f"User-Item matrix shape: {user_item_matrix.shape}")
print(f"Sparsity: {100 * (1 - user_item_matrix.nnz / (user_item_matrix.shape[0] * user_item_matrix.shape[1])):.2f}%")
```

```python
# Snippet 5: Implement Jaccard similarity for users and items
def jaccard_similarity_binary(matrix):
    """Calculate Jaccard similarity between rows (users)"""
    from sklearn.metrics.pairwise import pairwise_distances
    
    # Convert to binary (1 if rating > 0 else 0)
    binary_matrix = (matrix > 0).astype(np.float32).toarray()
    
    # Jaccard distance = 1 - Jaccard similarity
    # Jaccard similarity = intersection/union
    jaccard_distances = pairwise_distances(binary_matrix, metric='jaccard')
    jaccard_similarities = 1 - jaccard_distances
    
    return jaccard_similarities

def create_similarity_graph(similarity_matrix, threshold=0.3):
    """Create graph from similarity matrix using threshold"""
    n = similarity_matrix.shape[0]
    G = nx.Graph()
    
    # Add nodes
    G.add_nodes_from(range(n))
    
    # Add edges for similarity above threshold
    for i in range(n):
        for j in range(i+1, n):
            if similarity_matrix[i, j] > threshold:
                G.add_edge(i, j, weight=similarity_matrix[i, j])
    
    return G

# Calculate user similarity
user_similarity = jaccard_similarity_binary(user_item_matrix)
print(f"User similarity matrix shape: {user_similarity.shape}")
print(f"Similarity statistics - Mean: {user_similarity.mean():.3f}, Std: {user_similarity.std():.3f}")
```

```python
# Snippet 6: Create item-item similarity matrix
def create_item_item_similarity(user_item_matrix):
    # Transpose to get items as rows
    item_matrix = user_item_matrix.T
    
    # Calculate Jaccard similarity for items
    item_similarity = jaccard_similarity_binary(item_matrix)
    
    return item_similarity

# Calculate item similarity
item_similarity = create_item_item_similarity(user_item_matrix)
print(f"Item similarity matrix shape: {item_similarity.shape}")

# Create item similarity graph
item_graph = create_similarity_graph(item_similarity, threshold=0.2)
print(f"Item graph nodes: {item_graph.number_of_nodes()}, edges: {item_graph.number_of_edges()}")
```

```python
# Snippet 7: Implement community detection algorithms
def detect_communities_louvain(G):
    """Louvain community detection"""
    if G.number_of_edges() == 0:
        return {node: 0 for node in G.nodes()}, 0
    
    partition = community_louvain.best_partition(G, weight='weight')
    modularity = community_louvain.modularity(partition, G, weight='weight')
    return partition, modularity

def detect_communities_leiden(G):
    """Leiden algorithm using igraph"""
    if G.number_of_edges() == 0:
        return {node: 0 for node in G.nodes()}, 0
    
    # Convert to igraph
    G_ig = ig.Graph.from_networkx(G)
    
    # Run Leiden
    partition = la.find_partition(G_ig, la.ModularityVertexPartition, weights='weight')
    
    # Convert back to dict
    partition_dict = {node: partition.membership[i] for i, node in enumerate(G.nodes())}
    
    # Calculate modularity
    modularity = G_ig.modularity(partition.membership, weights=G_ig.es['weight'])
    
    return partition_dict, modularity

def detect_communities_infomap(G):
    """Infomap community detection using cdlib"""
    if G.number_of_edges() == 0:
        return {node: 0 for node in G.nodes()}, 0
    
    from cdlib import algorithms
    communities = algorithms.infomap(G)
    
    # Convert to partition dict
    partition_dict = {}
    for comm_id, nodes in enumerate(communities.communities):
        for node in nodes:
            partition_dict[node] = comm_id
    
    # Calculate modularity
    modularity = community_louvain.modularity(partition_dict, G, weight='weight')
    
    return partition_dict, modularity

def detect_communities_fluidc(G, n_communities=10):
    """Fluid Communities algorithm (requires connected graph)"""
    if G.number_of_edges() == 0 or nx.number_connected_components(G) > n_communities:
        return {node: 0 for node in G.nodes()}, 0
    
    try:
        from cdlib import algorithms
        communities = algorithms.fluid_communities(G, n_communities)
        
        # Convert to partition dict
        partition_dict = {}
        for comm_id, nodes in enumerate(communities.communities):
            for node in nodes:
                partition_dict[node] = comm_id
        
        modularity = community_louvain.modularity(partition_dict, G, weight='weight')
        return partition_dict, modularity
    except:
        return {node: 0 for node in G.nodes()}, 0
```

```python
# Snippet 8: Generate recommendations using communities
class CommunityRecommender:
    def __init__(self, similarity_matrix, partition, users_or_items, mode='user'):
        self.similarity_matrix = similarity_matrix
        self.partition = partition
        self.mode = mode
        self.data = users_or_items
        
        # Create community to indices mapping
        self.community_members = defaultdict(list)
        for idx, comm in partition.items():
            self.community_members[comm].append(idx)
    
    def recommend_user_based(self, user_idx, user_item_matrix, top_n=5):
        """Recommend items from same community users"""
        if user_idx not in self.partition:
            return []
        
        user_comm = self.partition[user_idx]
        community_users = [u for u in self.community_members[user_comm] if u != user_idx]
        
        if not community_users:
            return []
        
        # Get items purchased by community users
        community_items = defaultdict(float)
        for other_user in community_users:
            other_items = user_item_matrix[other_user].nonzero()[1]
            for idx in other_items:
                # Use similarity as weight
                weight = self.similarity_matrix[user_idx, other_user]
                community_items[idx] += weight
        
        # Get user's already purchased items
        user_items = set(user_item_matrix[user_idx].nonzero()[1])
        
        # Filter out items user already has
        recommendations = [(idx, score) for idx, score in community_items.items() 
                          if idx not in user_items]
        recommendations.sort(key=lambda x: x[1], reverse=True)
        
        return recommendations[:top_n]
    
    def recommend_item_based(self, item_idx, user_purchased_items=None, top_n=5):
        """Recommend items from same community"""
        if item_idx not in self.partition:
            return []
        
        item_comm = self.partition[item_idx]
        community_items = [i for i in self.community_members[item_comm] if i != item_idx]
        
        if not community_items:
            return []
        
        # Sort by similarity
        recommendations = [(idx, self.similarity_matrix[item_idx, idx]) 
                          for idx in community_items]
        recommendations.sort(key=lambda x: x[1], reverse=True)
        
        if user_purchased_items:
            recommendations = [(idx, score) for idx, score in recommendations 
                             if idx not in user_purchased_items]
        
        return recommendations[:top_n]
```

```python
# Snippet 9: Evaluate recommender system performance
def evaluate_recommender(recommender, user_item_matrix, test_items, mode='user', k_values=[5,10]):
    """Calculate precision, recall, and F1 at k"""
    results = {}
    
    for k in k_values:
        precisions = []
        recalls = []
        
        for user_idx in range(min(100, user_item_matrix.shape[0])):
            # Get user's items
            user_items = set(user_item_matrix[user_idx].nonzero()[1])
            
            if len(user_items) < 2:
                continue
            
            # Split into train/test
            train_items = set(np.random.choice(list(user_items), size=max(1, len(user_items)//2), replace=False))
            test_items_user = user_items - train_items
            
            if not test_items_user:
                continue
            
            if mode == 'user':
                recs = recommender.recommend_user_based(user_idx, user_item_matrix, top_n=k)
            else:
                # For item-based, recommend based on a random purchased item
                purchased_item = np.random.choice(list(train_items))
                recs = recommender.recommend_item_based(purchased_item, train_items, top_n=k)
            
            if recs:
                recommended_items = set([idx for idx, _ in recs])
                hits = len(recommended_items & test_items_user)
                
                precision = hits / len(recommended_items) if recommended_items else 0
                recall = hits / len(test_items_user) if test_items_user else 0
                
                precisions.append(precision)
                recalls.append(recall)
        
        if precisions and recalls:
            avg_precision = np.mean(precisions)
            avg_recall = np.mean(recalls)
            f1 = 2 * (avg_precision * avg_recall) / (avg_precision + avg_recall) if (avg_precision + avg_recall) > 0 else 0
            
            results[f'precision@{k}'] = avg_precision
            results[f'recall@{k}'] = avg_recall
            results[f'f1@{k}'] = f1
    
    return results
```

```python
# Snippet 10: Run user-user community detection and evaluation
# Create user similarity graph
user_graph = create_similarity_graph(user_similarity, threshold=0.3)
print(f"User graph: {user_graph.number_of_nodes()} nodes, {user_graph.number_of_edges()} edges")

# Run different community detection algorithms
algorithms = {
    'Louvain': detect_communities_louvain,
    'Leiden': detect_communities_leiden,
    'Infomap': detect_communities_infomap,
    'FluidC': lambda G: detect_communities_fluidc(G, n_communities=10)
}

user_results = {}

for algo_name, algo_func in algorithms.items():
    print(f"\n--- Running {algo_name} on User-User graph ---")
    partition, modularity = algo_func(user_graph)
    user_results[algo_name] = {'modularity': modularity, 'n_communities': len(set(partition.values()))}
    
    print(f"Modularity: {modularity:.4f}")
    print(f"Number of communities: {len(set(partition.values()))}")
    
    # Create recommender
    if modularity > 0:
        recommender = CommunityRecommender(user_similarity, partition, users, mode='user')
        
        # Evaluate
        eval_results = evaluate_recommender(recommender, user_item_matrix, None, mode='user', k_values=[5,10])
        user_results[algo_name].update(eval_results)
        
        print(f"Precision@5: {eval_results.get('precision@5', 0):.4f}")
        print(f"Recall@5: {eval_results.get('recall@5', 0):.4f}")
        print(f"F1@5: {eval_results.get('f1@5', 0):.4f}")

# Display results
results_df = pd.DataFrame(user_results).T
print("\n=== User-User Recommender Results ===")
print(results_df)
```

```python
# Snippet 11: Run item-item community detection and evaluation
# Create item similarity graph
item_graph = create_similarity_graph(item_similarity, threshold=0.2)
print(f"Item graph: {item_graph.number_of_nodes()} nodes, {item_graph.number_of_edges()} edges")

# For item-based, we need to run community detection
item_results = {}

for algo_name, algo_func in algorithms.items():
    print(f"\n--- Running {algo_name} on Item-Item graph ---")
    partition, modularity = algo_func(item_graph)
    item_results[algo_name] = {'modularity': modularity, 'n_communities': len(set(partition.values()))}
    
    print(f"Modularity: {modularity:.4f}")
    print(f"Number of communities: {len(set(partition.values()))}")
    
    # Create recommender
    if modularity > 0:
        recommender = CommunityRecommender(item_similarity, partition, items, mode='item')
        
        # Evaluate
        eval_results = evaluate_recommender(recommender, user_item_matrix, None, mode='item', k_values=[5,10])
        item_results[algo_name].update(eval_results)
        
        print(f"Precision@5: {eval_results.get('precision@5', 0):.4f}")
        print(f"Recall@5: {eval_results.get('recall@5', 0):.4f}")
        print(f"F1@5: {eval_results.get('f1@5', 0):.4f}")

# Display results
item_results_df = pd.DataFrame(item_results).T
print("\n=== Item-Item Recommender Results ===")
print(item_results_df)
```

```python
# Snippet 12: Compare both approaches
# Create comparison dataframe
comparison = pd.DataFrame({
    'Metric': ['Modularity', 'Precision@5', 'Recall@5', 'F1@5'],
    'User-User Best': [
        max(user_results.values(), key=lambda x: x.get('modularity', 0)).get('modularity', 0),
        max(user_results.values(), key=lambda x: x.get('precision@5', 0)).get('precision@5', 0),
        max(user_results.values(), key=lambda x: x.get('recall@5', 0)).get('recall@5', 0),
        max(user_results.values(), key=lambda x: x.get('f1@5', 0)).get('f1@5', 0)
    ],
    'Item-Item Best': [
        max(item_results.values(), key=lambda x: x.get('modularity', 0)).get('modularity', 0),
        max(item_results.values(), key=lambda x: x.get('precision@5', 0)).get('precision@5', 0),
        max(item_results.values(), key=lambda x: x.get('recall@5', 0)).get('recall@5', 0),
        max(item_results.values(), key=lambda x: x.get('f1@5', 0)).get('f1@5', 0)
    ]
})

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Modularity comparison
algorithms_names = list(user_results.keys())
user_mod = [user_results[algo]['modularity'] for algo in algorithms_names]
item_mod = [item_results[algo]['modularity'] for algo in algorithms_names]

x = np.arange(len(algorithms_names))
width = 0.35

axes[0].bar(x - width/2, user_mod, width, label='User-User', color='skyblue')
axes[0].bar(x + width/2, item_mod, width, label='Item-Item', color='lightcoral')
axes[0].set_xlabel('Algorithm')
axes[0].set_ylabel('Modularity')
axes[0].set_title('Modularity Comparison')
axes[0].set_xticks(x)
axes[0].set_xticklabels(algorithms_names, rotation=45)
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot 2: F1 Score comparison
user_f1 = [user_results[algo].get('f1@5', 0) for algo in algorithms_names]
item_f1 = [item_results[algo].get('f1@5', 0) for algo in algorithms_names]

axes[1].bar(x - width/2, user_f1, width, label='User-User', color='skyblue')
axes[1].bar(x + width/2, item_f1, width, label='Item-Item', color='lightcoral')
axes[1].set_xlabel('Algorithm')
axes[1].set_ylabel('F1 Score @5')
axes[1].set_title('F1 Score Comparison')
axes[1].set_xticks(x)
axes[1].set_xticklabels(algorithms_names, rotation=45)
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n=== Summary Table ===")
print(comparison.to_string(index=False))
print("\nBest performing approach:")
best_f1_user = max([user_results[algo].get('f1@5', 0) for algo in algorithms_names])
best_f1_item = max([item_results[algo].get('f1@5', 0) for algo in algorithms_names])

if best_f1_user > best_f1_item:
    print(f"User-User approach performs better (F1@5: {best_f1_user:.4f} vs {best_f1_item:.4f})")
else:
    print(f"Item-Item approach performs better (F1@5: {best_f1_item:.4f} vs {best_f1_user:.4f})")
```
